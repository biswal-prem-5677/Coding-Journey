```python
a = {'A': 10, 'B': 20}
b = {'B': 30, 'C': 40, 'D': 50}
df1 = pd.DataFrame(a, index=[0])
df2 = pd.DataFrame(b, index=[1])

# Corrected usage for d1 using pd.concat instead of the removed .append()
d1 = pd.concat([df1, df2]).fillna(0)
print(d1)

d2= pd.concat([df1, df2]).fillna(0)
print(d2)

d3= pd.DataFrame()
d3= d3.combine_first(df1).combine_first(df2).fillna(0)
print(d3)
```
