# **Python Dictionary**

Python dictionary is a data structure that stores the value in key: value pairs. Values in a dictionary can be of any data type and can be duplicated, whereas keys can't be repeated and must be immutable.


*   Keys are case sensitive which means same name but different cases of Key will be treated distinctly.

*   Keys must be immutable which means keys can be strings, numbers or tuples but not lists.
*   Duplicate keys are not allowed and any duplicate key will overwrite the previous value.


*   Internally uses hashing. Hence, operations like search, insert, delete can be performed in Constant Time.

*   From Python 3.7 Version onward, Python dictionary are Ordered.


# **How to Create a Dictionary**
Dictionary can be created by placing a sequence of elements within curly {} braces, separated by a 'comma'.

```python
d1 = {1: 'Geeks', 2: 'For', 3: 'Geeks'}
print(d1)

# create dictionary using dict() constructor
d2 = dict(a = "Geeks", b = "for", c = "Geeks")
print(d2)
```

```python
dict = {'Vintage': 'Ambassador', 'Luxury': 'Rolls Royce', 'Sport': 'Buggati'}
print(dict)
```

Accessing Dictionary Items
We can access a value from a dictionary by using the key within square brackets or [get()](https://www.geeksforgeeks.org/python/python-dictionary-get-method/) method.





```python
d = { "name": "Prem", 1: "Python", (1, 2): [1,2,4] }

# Access using key
print(d["name"])

# Access using get()
print(d.get("name"))
```

```python
dict = {'Vintage': 'Ambassador', 'Luxury': 'Rolls Royce', 'Sport': 'Buggati'}
print(dict['Luxury'])
print(dict.get('Sport'))
```

```python
for x in dict:
  print(x)
```

```python
for x in dict.values():
  print(x)
```

```python
for x, y in dict.items():
  print(x, y)
```

# **Adding and Updating Dictionary Items**
We can add new key-value pairs or update existing keys by using assignment.

```python
d = {1: 'Geeks', 2: 'For', 3: 'Geeks'}

# Adding a new key-value pair
d["age"] = 22

# Updating an existing value
d[1] = "Python dict"

print(d)
```

```python
dict = {'Vintage': 'Ambassador', 'Luxury': 'Rolls Royce', 'Sport': 'Buggati'}
dict['Luxury'] = 'Mercedes'
dict['SUV'] = 'BMW'
print(dict)
```

```python
for type, choice in dict.items():
  print(type, choice)
```

# **Removing Dictionary Items**
We can remove items from dictionary using the following methods:


1.   del: Removes an item by key.

1.   pop(): Removes an item by key and returns its value.
2.   clear(): Empties the dictionary.


2.   popitem(): Removes and returns the last key-value pair.







```python
dict = {'Vintage': 'Ambassador', 'Luxury': 'Rolls Royce', 'Sport': 'Buggati'}
del dict['Vintage']
print(dict)
```

```python
dict = {'Vintage': 'Ambassador', 'Luxury': 'Rolls Royce', 'Sport': 'Buggati'}
dict.pop('Vintage')
```

```python
dict = {'Vintage': 'Ambassador', 'Luxury': 'Rolls Royce', 'Sport': 'Buggati'}
dict.popitem()
```

```python
dict = {'Vintage': 'Ambassador', 'Luxury': 'Rolls Royce', 'Sport': 'Buggati'}
dict.clear()
print(dict)
```
