```python
# ============================================
# LEVEL 1: STORE DOCUMENTS
# ============================================

documents = [
    "AI is transforming the world",
    "Machine learning is a subset of AI",
    "Deep learning uses neural networks",
    "Python is widely used in AI"
]
```

```python
from transformers import AutoTokenizer, AutoModel
import torch

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")

def embed(text):
    inputs = tokenizer(text, return_tensors="pt")
    outputs = model(**inputs)
    return outputs.last_hidden_state.mean(dim=1)
```

```python
doc_embeddings = [embed(doc) for doc in documents]
```

```python
import torch.nn.functional as F

query = "What is AI?"

query_emb = embed(query)

scores = []

for emb in doc_embeddings:
    score = F.cosine_similarity(query_emb, emb)
    scores.append(score.item())

best_idx = scores.index(max(scores))

retrieved_doc = documents[best_idx]

print("Retrieved:", retrieved_doc)
```

```python
# Simulated generation (LLM-style)
answer = f"Based on knowledge: {retrieved_doc}"

print("Final Answer:", answer)
```

```python
top_k = sorted(range(len(scores)), key=lambda i: scores[i], reverse=True)[:2]

context = " ".join([documents[i] for i in top_k])

print("Context:", context)
```
