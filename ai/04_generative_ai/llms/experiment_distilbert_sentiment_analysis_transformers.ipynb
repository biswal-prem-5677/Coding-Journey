## Text Classification with Pre-trained Models

Text classification is the task of assigning labels or categories to text. For example, categorizing emails as 'spam' or 'not spam', or classifying movie reviews as 'positive' or 'negative'.

Pre-trained models, especially large language models, have been trained on vast amounts of text data and have learned general language understanding. We can then *fine-tune* these models on a specific text classification task, or in many cases, use models that have already been fine-tuned for a common task.

Here, I'll demonstrate using a pre-trained and fine-tuned model from Hugging Face's `transformers` library for sentiment analysis (classifying text as positive or negative).

Now, let's import the necessary components from the `transformers` library. We'll use `pipeline` for a high-level API to perform text classification easily, and `AutoTokenizer` and `AutoModelForSequenceClassification` if we want more control or to load specific models.

````python
from transformers import pipeline

# Load a pre-trained sentiment analysis model
# This model is fine-tuned on the Stanford Sentiment Treebank (SST-2) dataset.
classifier = pipeline('sentiment-analysis', model="distilbert-base-uncased-finetuned-sst-2-english")

# Example texts for classification
texts = [
    "I love using pre-trained models, they make AI development so much faster!",
    "This movie was absolutely dreadful and a waste of time.",
    "The weather is neither good nor bad today, just typical."
]

# Perform classification
results = classifier(texts)

# Display the results
for i, text in enumerate(texts):
    print(f"Text: '{text}'")
    print(f"Classification: {results[i]['label']} (Score: {results[i]['score']:.4f})\n")
````

The `pipeline` function abstracts away much of the complexity. If you wanted more granular control, you would typically load the tokenizer and model separately:

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")
model = AutoModelForSequenceClassification.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")

inputs = tokenizer(texts[0], return_tensors="pt") # Tokenize the first example text
outputs = model(**inputs)

# The output are logits, which can be converted to probabilities using softmax
probabilities = torch.softmax(outputs.logits, dim=1)

# Get the predicted label (index of the max probability)
predicted_class_id = probabilities.argmax().item()
predicted_label = model.config.id2label[predicted_class_id]

print(f"Text: '{texts[0]}'\nPredicted Label: {predicted_label} (Probability: {probabilities.max().item():.4f})")
```

This gives you a more detailed understanding of the model's output, especially useful if you need to access raw logits or probabilities for more complex tasks or custom post-processing.

````python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")
model = AutoModelForSequenceClassification.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")

inputs = tokenizer(texts[0], return_tensors="pt") # Tokenize the first example text
outputs = model(**inputs)

# The output are logits, which can be converted to probabilities using softmax
probabilities = torch.softmax(outputs.logits, dim=1)

# Get the predicted label (index of the max probability)
predicted_class_id = probabilities.argmax().item()
predicted_label = model.config.id2label[predicted_class_id]

print(f"Text: '{texts[0]}'\nPredicted Label: {predicted_label} (Probability: {probabilities.max().item():.4f})")
````
