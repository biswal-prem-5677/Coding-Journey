```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
s
model_name = "google/flan-t5-base"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
def generate_text(prompt, max_tokens=128):
    inputs = tokenizer(prompt, return_tensors="pt")
    outputs = model.generate(**inputs, max_new_tokens=max_tokens)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
def generate_excuse(context, urgency):
    prompt = f"Give a realistic and creative excuse for: {context}. Urgency level: {urgency}."
    return generate_text(prompt)

def generate_apology(context):
    prompt = f"Write a short, sincere apology for this situation: {context}."
    return generate_text(prompt)

def generate_fake_proof(context):
    prompt = f"Create a fake believable proof (text or note) for this excuse: {context}."
    return generate_text(prompt)
```

```python
context = input("Enter situation/context (e.g., 'missed school project deadline'): ")
urgency = input("Enter urgency (low/medium/high): ")

excuse = generate_excuse(context, urgency)
apology = generate_apology(context)
proof = generate_fake_proof(context)

print("\n--- Excuse ---\n", excuse)
print("\n--- Apology ---\n", apology)
print("\n--- Fake Proof ---\n", proof)
```
