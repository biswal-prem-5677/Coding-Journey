```python
# ============================================
# LEVEL 1: ADD NOISE
# ============================================

import torch
import matplotlib.pyplot as plt

# Fake "image"
x = torch.randn(1, 100)

noise = torch.randn_like(x)

# Add noise
noisy = x + 0.5 * noise

plt.plot(x.numpy()[0], label="Original")
plt.plot(noisy.numpy()[0], label="Noisy")

plt.legend()
plt.title("Forward Diffusion")
plt.show()
```

```python
# ============================================
# MULTI-STEP NOISE
# ============================================

x = torch.randn(1, 100)

steps = 5
data = [x]

for i in range(steps):
    noise = torch.randn_like(x)
    x = x + 0.5 * noise
    data.append(x)

# Plot progression
for i, d in enumerate(data):
    plt.plot(d.numpy()[0], label=f"Step {i}")

plt.legend()
plt.title("Noise Progression")
plt.show()
```

```python
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(100, 128),
    nn.ReLU(),
    nn.Linear(128, 100)
)

# Training idea:
# loss = predicted_noise - actual_noise
```

```python
x = torch.randn(1, 100)  # start from noise

for step in range(5):
    predicted_noise = model(x)
    x = x - predicted_noise * 0.1  # denoise

print("Generated:", x.shape)
```
