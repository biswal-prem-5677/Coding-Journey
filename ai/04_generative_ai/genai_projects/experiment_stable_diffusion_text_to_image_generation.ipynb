## Generate AI Images with Stable Diffusion using Diffusers

This section will demonstrate how to use the `diffusers` library to generate images from text prompts using a Stable Diffusion model.

```python
# Install necessary libraries (excluding diffusers for now, will install from source)
!pip install -qq transformers accelerate torch
```

```python
import torch
from diffusers import DiffusionPipeline

# Check for GPU availability
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")
```

```python
# Load a pre-trained Stable Diffusion model
# This might take some time to download the model weights (~2.5 GB)
pipeline = DiffusionPipeline.from_pretrained("runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16)
pipeline.to(device)
```

```python
# Define your text prompt
prompt = "a photo of a warrior with sward in hand riding a horse on mars, detailed, vibrant, high resolution"
```

```python
# Generate the image
# You can adjust `num_inference_steps` for quality vs. speed trade-off
image = pipeline(prompt, num_inference_steps=50).images[0]

# Display the generated image
display(image)
```
