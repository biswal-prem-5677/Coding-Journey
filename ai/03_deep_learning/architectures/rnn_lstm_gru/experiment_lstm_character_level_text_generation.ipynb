```python
import  numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM
```

```python
text="""This is the first sentence.
This is the second sentence.
And this is the third one."""
```

```python
text=text.lower()
```

```python
chars=sorted(list(set(text)))
```

```python
char_to_idx={c:i for i,c in enumerate(chars)}
idx_to_char={i:c for i,c in enumerate(chars)}
```

```python
vocab_size=len(chars)
```

```python
seq_length = 10
X = []
y = []

for i in range(len(text) - seq_length):
    seq = text[i:i + seq_length]
    target = text[i + seq_length]

    X.append([char_to_idx[char] for char in seq])
    y.append(char_to_idx[target])

X = np.array(X)
y = np.array(y)

#Normalize
X=X/float(vocab_size)

# reshape for LSTM
X = X.reshape((X.shape[0], seq_length,1))
```

```python
model = Sequential()
model.add(LSTM(128, input_shape=(seq_length, 1)))
model.add(Dense(vocab_size, activation='softmax'))

model.compile(loss='sparse_categorical_crossentropy', optimizer='adam')
model.summary()
```

```python
#training
model.fit(X,y,epochs=200, batch_size=32)
```

```python
start = text[0:seq_length]
generated = start
print("\nStarting Seed: ",start)
for i in range(200):
  input_seq = [char_to_idx[c] for c in start]
  input_seq = np.array(input_seq)/float(vocab_size)
  input_seq = input_seq.reshape((1, seq_length, 1))
  prediction = model.predict(input_seq, verbose=0)
  #predicted_char = idx_to_char[np.argmax(prediction)]

  #Random sampling
  index = np.random.choice(len(prediction[0]), p=prediction[0])
  next_char = idx_to_char[index]

  generated += next_char
  start = start[1:] + next_char

print("\nGenerated Text:\n ",generated)
```
