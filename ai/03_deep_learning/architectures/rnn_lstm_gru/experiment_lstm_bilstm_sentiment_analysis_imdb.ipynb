```python
# ============================================
# Sentiment Analysis using LSTM (Deep Learning)
# Clean, Optimized, Readable Code
# ============================================

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.datasets import imdb

# -------------------------------
# 1. Load Dataset (IMDB Reviews)
# -------------------------------
VOCAB_SIZE = 10000   # top 10k words
MAX_LEN = 200        # max review length

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=VOCAB_SIZE)

# -------------------------------
# 2. Preprocessing (Padding)
# -------------------------------
x_train = pad_sequences(x_train, maxlen=MAX_LEN, padding='post')
x_test = pad_sequences(x_test, maxlen=MAX_LEN, padding='post')

# -------------------------------
# 3. Build LSTM Model
# -------------------------------
model = models.Sequential([

    # Convert word indices → dense vectors
    layers.Embedding(input_dim=VOCAB_SIZE, output_dim=128, input_length=MAX_LEN),

    # LSTM layer (captures sequence context)
    layers.LSTM(128, return_sequences=False),

    # Regularization
    layers.Dropout(0.5),

    # Fully connected layer
    layers.Dense(64, activation='relu'),

    # Output layer (binary classification)
    layers.Dense(1, activation='sigmoid')
])

# -------------------------------
# 4. Compile Model
# -------------------------------
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# -------------------------------
# 5. Train Model
# -------------------------------
history = model.fit(
    x_train, y_train,
    epochs=5,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)

# -------------------------------
# 6. Evaluate Model
# -------------------------------
test_loss, test_acc = model.evaluate(x_test, y_test)

print(f"\n✅ Test Accuracy: {test_acc:.4f}")
```

```python
# ============================================
# Advanced Sentiment Analysis using BiLSTM
# Better Accuracy + Cleaner + More Powerful
# ============================================

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.datasets import imdb

# -------------------------------
# 1. Load Dataset
# -------------------------------
VOCAB_SIZE = 10000
MAX_LEN = 200

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=VOCAB_SIZE)

# -------------------------------
# 2. Padding Sequences
# -------------------------------
x_train = pad_sequences(x_train, maxlen=MAX_LEN, padding='post')
x_test = pad_sequences(x_test, maxlen=MAX_LEN, padding='post')

# -------------------------------
# 3. Build Improved Model
# -------------------------------
model = models.Sequential([

    # Embedding Layer (better representation)
    layers.Embedding(VOCAB_SIZE, 128, input_length=MAX_LEN),

    # Bidirectional LSTM (captures past + future context)
    layers.Bidirectional(layers.LSTM(128, return_sequences=True)),

    # Second LSTM layer (deeper understanding)
    layers.LSTM(64),

    # Regularization
    layers.Dropout(0.5),

    # Dense Layer
    layers.Dense(64, activation='relu'),

    # Output Layer
    layers.Dense(1, activation='sigmoid')
])

# -------------------------------
# 4. Compile
# -------------------------------
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# -------------------------------
# 5. Callbacks (Smart Training)
# -------------------------------
early_stop = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=2,
    restore_best_weights=True
)

lr_reduce = tf.keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.3,
    patience=1
)

# -------------------------------
# 6. Train
# -------------------------------
history = model.fit(
    x_train, y_train,
    epochs=8,
    batch_size=128,
    validation_split=0.2,
    callbacks=[early_stop, lr_reduce],
    verbose=1
)

# -------------------------------
# 7. Evaluate
# -------------------------------
test_loss, test_acc = model.evaluate(x_test, y_test)

print(f"\n🔥 Test Accuracy: {test_acc:.4f}")
```
