```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])

y = np.array([0,1,1,0])
```

```python
# Build Model
model = keras.Sequential([
    keras.layers.Dense(8, activation='relu', input_shape=(2,)), #Dense(8) Hidden layer with 8 neurons
    keras.layers.Dense(1, activation='sigmoid')
])
```

```python
# Compile
model.compile(optimizer='adam',                                 #Updates weights efficiently
              loss='binary_crossentropy',                       #Measures classification error
              metrics=['accuracy'])
```

```python
# Train
history = model.fit(X, y, epochs=200, verbose=1)                 #No training output displayed
```

```python
# Evaluate
loss, accuracy = model.evaluate(X, y)
print("Accuracy:", accuracy)

# ----------------------------
# Plot Loss and Accuracy
# ----------------------------

plt.figure()
plt.plot(history.history['loss'])
plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
plt.figure()
plt.plot(history.history['accuracy'])
plt.title("Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.show()
```

```python
x1 = float(input("Enter value for x1 (0 or 1): "))
x2 = float(input("Enter value for x2 (0 or 1): "))

user_input = np.array([[x1, x2]])

prediction = model.predict(user_input)

print("Raw Output:", prediction[0][0])

if prediction[0][0] >= 0.5:
    print("Prediction (Neural Network): 1")
else:
    print("Prediction (Neural Network): 0")
```
