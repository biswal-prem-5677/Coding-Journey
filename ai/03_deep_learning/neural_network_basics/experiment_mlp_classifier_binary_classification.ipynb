```python
from sklearn.neural_network import MLPClassifier
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split

# Generate some synthetic data
X, y = make_classification(n_samples=1000, n_features=10, n_informative=5, n_redundant=0, random_state=42)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, random_state=42)

# Create an MLPClassifier (Multi-layer Perceptron Classifier) model
# This is a basic feedforward ANN
# hidden_layer_sizes=(100,) means one hidden layer with 100 neurons
# max_iter is the maximum number of iterations (epochs)
mlp = MLPClassifier(hidden_layer_sizes=(100,), max_iter=1000, activation='relu', solver='adam', random_state=42, tol=1e-4) # Increased max_iter to 1000 and added tol

# Train the model
mlp.fit(X_train, y_train)

# Evaluate the model
accuracy = mlp.score(X_test, y_test)
print(f"Model accuracy: {accuracy:.4f}")

# Make predictions on new data
# For demonstration, let's predict on the first 5 test samples
predictions = mlp.predict(X_test[:5])
print(f"\nPredictions for first 5 test samples: {predictions}")
print(f"Actual labels for first 5 test samples: {y_test[:5]}")
```
