```python
import os # import the 'os' module
import nbformat as nbf
from nbformat import write
from nbformat.v4 import new_notebook # This imports new_notebook from nbformat.v4

# Define base_path
base_path = r"/content/C:\Users\priya\OneDrive\เอกสาร\internship project/Case_Study_Healthcare"   #  desired base directory

# Create and populate the Case Study - Healthcare notebook
case_study_path = os.path.join(base_path, "Case_Study_Healthcare")
os.makedirs(case_study_path, exist_ok=True)
file_path = os.path.join(case_study_path, "Case_Study_Healthcare.ipynb")

nb = new_notebook() # Call new_notebook directly as it was imported from nbformat.v4

# Add content for Healthcare Case Study
cells = [
    nbf.v4.new_markdown_cell("## 🩺 Case Study: Artificial Intelligence in Healthcare\n\n"
                          "In this case study, we explore the applications, challenges, and future scope of AI in the healthcare industry."),

    nbf.v4.new_markdown_cell("### 📌 Introduction\n"
                          "Healthcare is undergoing a technological revolution driven by Artificial Intelligence (AI). "
                          "From diagnostics and treatment planning to patient monitoring and drug discovery, AI is enabling faster, more accurate, and personalized care."),

    nbf.v4.new_markdown_cell("### 🔍 Applications of AI in Healthcare\n"
                          "- **Disease Diagnosis**: AI models help detect diseases like cancer, diabetes, and cardiovascular conditions using imaging and patient data.\n"
                          "- **Medical Imaging**: Deep learning techniques are used in X-rays, MRIs, and CT scans for accurate interpretation.\n"
                          "- **Virtual Health Assistants**: Chatbots and virtual nurses assist in patient triage, medication reminders, and symptom checking.\n"
                          "- **Predictive Analytics**: AI predicts disease outbreaks, patient readmissions, and personalized treatment outcomes.\n"
                          "- **Robotic Surgery**: AI-powered robots assist surgeons with precision and minimal invasiveness.\n"
                          "- **Drug Discovery**: AI accelerates the process of identifying and testing new compounds for drug development."),

    nbf.v4.new_markdown_cell("### 🧠 Real-world Examples\n"
                          "- **IBM Watson Health**: Supports clinical decision-making and evidence-based treatment.\n"
                          "- **Google DeepMind**: Accurately detects over 50 eye diseases from 3D retinal scans.\n"
                          "- **Aidoc**: AI solution for radiologists to identify acute medical conditions in CT scans."),

    nbf.v4.new_markdown_cell("### ⚙️ Technologies Used\n"
                          "- Machine Learning (ML), Deep Learning (CNNs, RNNs)\n"
                          "- Natural Language Processing (NLP)\n"
                          "- Computer Vision\n"
                          "- Internet of Medical Things (IoMT)\n"
                          "- Cloud Computing & Edge AI"),

    nbf.v4.new_markdown_cell("### 🚧 Challenges in AI Healthcare\n"
                          "- Data privacy and HIPAA compliance\n"
                          "- Lack of labeled data and data standardization\n"
                          "- Algorithm transparency and explainability\n"
                          "- Integration with legacy systems\n"
                          "- Ethical concerns and bias in training data"),

    nbf.v4.new_markdown_cell("### 🌐 Future Scope\n"
                          "- AI-powered personalized medicine\n"
                          "- Smart hospitals and real-time monitoring\n"
                          "- Integration with genomics and wearable tech\n"
                          "- AI in mental health diagnosis and support\n"
                          "- Global pandemic preparedness through AI surveillance"),

    nbf.v4.new_markdown_cell("### ✅ Conclusion\n"
                          "AI is transforming healthcare by making it more predictive, preventive, and personalized. "
                          "With continued innovation and ethical implementation, AI has the potential to enhance outcomes for billions of lives worldwide.")
]

# Add cells and save the notebook
nb.cells.extend(cells)
print(file_path) # Print the file path for verification
try:
    with open(file_path, 'w') as f:
        write(nb, f)
except Exception as e:
    print(f"An error occurred: {e}")

file_path
```
