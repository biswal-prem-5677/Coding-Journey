```python
import cv2
from google.colab.patches import cv2_imshow

# Read image
image = cv2.imread('/content/drive/MyDrive/eagle 3.jpg')

# Display image
cv2_imshow(image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python

import cv2
from google.colab.patches import cv2_imshow

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

cv2_imshow(gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```
