```python
# ============================================
# BERNOULLI DISTRIBUTION
# ============================================

import numpy as np
import matplotlib.pyplot as plt

p = 0.7  # probability of success

samples = np.random.binomial(1, p, 1000)

values, counts = np.unique(samples, return_counts=True)
probs = counts / len(samples)

print("Values:", values)
print("Probabilities:", probs)
```

```python
plt.bar(values, probs)
plt.title("Bernoulli Distribution")
plt.show()
```

```python
# ============================================
# BINOMIAL DISTRIBUTION
# ============================================

n = 10   # trials
p = 0.5  # probability

samples = np.random.binomial(n, p, 10000)

plt.hist(samples, bins=10)
plt.title("Binomial Distribution")
plt.xlabel("Number of successes")
plt.show()
```

```python
# ============================================
# NORMAL DISTRIBUTION
# ============================================

mean = 0
std = 1

data = np.random.normal(mean, std, 10000)

plt.hist(data, bins=50)
plt.title("Normal Distribution")
plt.show()
```

```python
plt.figure()

# Bernoulli
bernoulli = np.random.binomial(1, 0.5, 1000)
plt.hist(bernoulli, bins=2, alpha=0.5, label="Bernoulli")

# Binomial
binomial = np.random.binomial(10, 0.5, 1000)
plt.hist(binomial, bins=10, alpha=0.5, label="Binomial")

# Normal
normal = np.random.normal(0, 1, 1000)
plt.hist(normal, bins=30, alpha=0.5, label="Normal")

plt.legend()
plt.title("Distribution Comparison")
plt.show()
```

```python
# ============================================
# UNIFORM DISTRIBUTION
# ============================================

import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(0, 1, 10000)

plt.hist(data, bins=50)
plt.title("Uniform Distribution")
plt.show()
```

```python
# ============================================
# POISSON DISTRIBUTION
# ============================================

lam = 3  # average events

data = np.random.poisson(lam, 10000)

plt.hist(data, bins=20)
plt.title("Poisson Distribution")
plt.show()
```

```python
# ============================================
# EXPONENTIAL DISTRIBUTION
# ============================================

data = np.random.exponential(scale=1, size=10000)

plt.hist(data, bins=50)
plt.title("Exponential Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,6))

# Uniform
plt.hist(np.random.uniform(0,1,1000), bins=30, alpha=0.5, label="Uniform")

# Poisson
plt.hist(np.random.poisson(3,1000), bins=20, alpha=0.5, label="Poisson")

# Normal
plt.hist(np.random.normal(0,1,1000), bins=30, alpha=0.5, label="Normal")

plt.legend()
plt.title("Key Distributions")
plt.show()
```

```python
# ============================================
# MASTER DISTRIBUTION COMPARISON
# ============================================

import numpy as np
import matplotlib.pyplot as plt

samples = 10000

# -------------------------------
# Generate Data
# -------------------------------
bernoulli = np.random.binomial(1, 0.5, samples)
binomial = np.random.binomial(10, 0.5, samples)
poisson = np.random.poisson(3, samples)
uniform = np.random.uniform(0, 1, samples)
normal = np.random.normal(0, 1, samples)
exponential = np.random.exponential(1, samples)

# -------------------------------
# Plot All
# -------------------------------
plt.figure(figsize=(12,10))

# Bernoulli
plt.subplot(3,2,1)
plt.hist(bernoulli, bins=2)
plt.title("Bernoulli")

# Binomial
plt.subplot(3,2,2)
plt.hist(binomial, bins=10)
plt.title("Binomial")

# Poisson
plt.subplot(3,2,3)
plt.hist(poisson, bins=20)
plt.title("Poisson")

# Uniform
plt.subplot(3,2,4)
plt.hist(uniform, bins=30)
plt.title("Uniform")

# Normal
plt.subplot(3,2,5)
plt.hist(normal, bins=30)
plt.title("Normal")

# Exponential
plt.subplot(3,2,6)
plt.hist(exponential, bins=30)
plt.title("Exponential")

plt.suptitle("All Distribution Comparison")
plt.tight_layout()
plt.show()
```

```python
# ============================================
# STATISTICS COMPARISON
# ============================================

datasets = {
    "Bernoulli": bernoulli,
    "Binomial": binomial,
    "Poisson": poisson,
    "Uniform": uniform,
    "Normal": normal,
    "Exponential": exponential
}

for name, data in datasets.items():
    print(f"{name}:")
    print("  Mean:", np.mean(data))
    print("  Variance:", np.var(data))
    print()
```

```python
from scipy.stats import skew

for name, data in datasets.items():
    print(f"{name} Skewness:", skew(data))
```

```python
plt.hist(normal, bins=50, alpha=0.5, label="Normal")
plt.hist(uniform, bins=50, alpha=0.5, label="Uniform")
plt.legend()
plt.title("Normal vs Uniform")
plt.show()
```
