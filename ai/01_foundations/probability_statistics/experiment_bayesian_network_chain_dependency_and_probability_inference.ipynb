```python
# ============================================
# LEVEL 1: SIMPLE BAYES NETWORK
# ============================================

# Probabilities
p_rain = 0.3

p_traffic_given_rain = 0.8
p_traffic_given_no_rain = 0.2

# Total probability of traffic
p_traffic = (p_traffic_given_rain * p_rain +
             p_traffic_given_no_rain * (1 - p_rain))

print("P(Traffic):", p_traffic)
```

```python
# ============================================
# LEVEL 2: CHAIN DEPENDENCY
# ============================================

p_late_given_traffic = 0.9
p_late_given_no_traffic = 0.1

p_late = (p_late_given_traffic * p_traffic +
          p_late_given_no_traffic * (1 - p_traffic))

print("P(Late):", p_late)
```

```python
# Joint probability (example case: Rain, Traffic, Late)

joint = p_rain * p_traffic_given_rain * p_late_given_traffic

print("Joint probability:", joint)
```

```python
p_rain_given_traffic = (p_traffic_given_rain * p_rain) / p_traffic

print("P(Rain | Traffic):", p_rain_given_traffic)
```

```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()

G.add_edges_from([
    ("Rain", "Traffic"),
    ("Traffic", "Late")
])

nx.draw(G, with_labels=True, node_size=2000)
plt.title("Bayesian Network")
plt.show()
```
