```python
# ============================================
# LEVEL 1: BAYES EXAMPLE
# ============================================

# Given
p_d = 0.01          # P(Disease)
p_pos_d = 0.95      # P(Positive | Disease)
p_pos_nd = 0.05     # P(Positive | No Disease)

# Total probability of positive
p_pos = p_pos_d * p_d + p_pos_nd * (1 - p_d)

# Bayes theorem
p_d_given_pos = (p_pos_d * p_d) / p_pos

print("P(Disease | Positive):", p_d_given_pos)
```

```python
import matplotlib.pyplot as plt

labels = ["Prior P(D)", "Posterior P(D|+)"]
values = [p_d, p_d_given_pos]

plt.bar(labels, values)
plt.title("Belief Update with Bayes")
plt.show()
```

```python
# ============================================
# LEVEL 3: SIMPLE BAYES CLASSIFIER
# ============================================

# Fake probabilities
p_spam = 0.4
p_word_given_spam = 0.8
p_word_given_not = 0.2

# Total probability
p_word = p_word_given_spam * p_spam + p_word_given_not * (1 - p_spam)

# Posterior
p_spam_given_word = (p_word_given_spam * p_spam) / p_word

print("P(Spam | Word):", p_spam_given_word)
```

```python
def bayes(prior, likelihood, evidence):
    return (likelihood * prior) / evidence

print(bayes(0.01, 0.95, p_pos))
```
