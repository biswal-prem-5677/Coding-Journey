```python
# ============================================
# LEVEL 1: COIN MLE
# ============================================

heads = 7
tails = 3

p_hat = heads / (heads + tails)

print("Estimated probability of head:", p_hat)
```

```python
import numpy as np
import matplotlib.pyplot as plt

p_values = np.linspace(0.01, 0.99, 100)

likelihoods = []

for p in p_values:
    likelihood = (p**heads) * ((1-p)**tails)
    likelihoods.append(likelihood)

plt.plot(p_values, likelihoods)
plt.title("Likelihood Function")
plt.xlabel("p")
plt.ylabel("Likelihood")
plt.show()
```

```python
log_likelihoods = []

for p in p_values:
    log_likelihood = heads*np.log(p) + tails*np.log(1-p)
    log_likelihoods.append(log_likelihood)

plt.plot(p_values, log_likelihoods)
plt.title("Log Likelihood")
plt.show()
```
