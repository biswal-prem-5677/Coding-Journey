```python
import numpy as np

# Assume posterior ~ normal (for demo)
samples = np.random.normal(2, 0.5, 1000)

print("Sampled parameters:", samples[:5])
```

```python
x = 1.5

predictions = samples * x  # y = θx

mean_pred = np.mean(predictions)
std_pred = np.std(predictions)

print("Prediction:", mean_pred)
print("Uncertainty:", std_pred)
```

```python
# Posterior over weights ~ Normal(mean, variance)

w_samples = np.random.normal(3, 0.3, 1000)

x = 2

preds = w_samples * x

print("Mean prediction:", np.mean(preds))
print("Uncertainty:", np.std(preds))
```

```python
import matplotlib.pyplot as plt

plt.hist(preds, bins=50)
plt.title("Prediction Distribution")
plt.show()
```
