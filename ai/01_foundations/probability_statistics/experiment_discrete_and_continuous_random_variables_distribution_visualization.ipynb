```python
# ============================================
# DISCRETE RANDOM VARIABLE (DICE)
# ============================================

import numpy as np
import matplotlib.pyplot as plt

# Simulate dice rolls
trials = 10000
dice = np.random.randint(1, 7, trials)

# Probability distribution
values, counts = np.unique(dice, return_counts=True)
probs = counts / trials

print("Values:", values)
print("Probabilities:", probs)
```

```python
plt.bar(values, probs)
plt.title("Discrete Distribution (Dice)")
plt.xlabel("Value")
plt.ylabel("Probability")
plt.show()
```

```python
# ============================================
# CONTINUOUS RANDOM VARIABLE
# ============================================

data = np.random.randn(10000)  # normal distribution

plt.hist(data, bins=50)
plt.title("Continuous Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
pmf = probs
print("PMF:", pmf)
```

```python
hist, bins = np.histogram(data, bins=50, density=True)

plt.plot(bins[:-1], hist)
plt.title("PDF Approximation")
plt.show()
```
