```python
# ============================================
# LEVEL 1: BASIC PROBABILITY (COIN SIMULATION)
# ============================================

import numpy as np
import matplotlib.pyplot as plt

# Number of experiments
trials = 1000

# Simulate coin toss (0 = tail, 1 = head)
results = np.random.randint(0, 2, trials)

# Count heads
heads = np.sum(results)

probability = heads / trials

print("Heads:", heads)
print("Estimated Probability:", probability)
```

```python
# Running probability (how it stabilizes)

running_prob = np.cumsum(results) / np.arange(1, trials+1)

plt.figure()
plt.plot(running_prob)
plt.axhline(0.5, linestyle='--')  # true probability

plt.title("Law of Large Numbers")
plt.xlabel("Trials")
plt.ylabel("Probability of Heads")

plt.show()
```

```python
# ============================================
# LEVEL 2: DICE SIMULATION
# ============================================

trials = 5000

dice = np.random.randint(1, 7, trials)

# Probability of getting 6
prob_6 = np.sum(dice == 6) / trials

print("P(6):", prob_6)
```

```python
plt.hist(dice, bins=6)
plt.title("Dice Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
even = dice[dice % 2 == 0]

prob_6_given_even = np.sum(even == 6) / len(even)

print("P(6 | even):", prob_6_given_even)
```

```python
# P(Disease) = 0.01
# P(Positive | Disease) = 0.95
# P(Positive | No Disease) = 0.05

p_d = 0.01
p_pos_d = 0.95
p_pos_nd = 0.05

# Bayes theorem
p_pos = p_pos_d * p_d + p_pos_nd * (1 - p_d)

p_d_given_pos = (p_pos_d * p_d) / p_pos

print("P(Disease | Positive):", p_d_given_pos)
```
