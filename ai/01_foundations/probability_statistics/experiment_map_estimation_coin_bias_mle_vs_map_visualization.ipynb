```python
# ============================================
# LEVEL 1: MAP (COIN)
# ============================================

heads = 7
tails = 3

# Prior (Beta distribution idea)
alpha = 2   # prior heads
beta = 2    # prior tails

# MAP estimate
p_map = (heads + alpha - 1) / (heads + tails + alpha + beta - 2)

print("MAP estimate:", p_map)
```

```python
import numpy as np
import matplotlib.pyplot as plt

p_values = np.linspace(0.01, 0.99, 100)

likelihood = [(p**heads)*((1-p)**tails) for p in p_values]

# Simple prior (favor middle values)
prior = [p*(1-p) for p in p_values]

posterior = np.array(likelihood) * np.array(prior)

plt.plot(p_values, likelihood, label="Likelihood (MLE)")
plt.plot(p_values, posterior, label="Posterior (MAP)")

plt.legend()
plt.title("MLE vs MAP")
plt.show()
```
