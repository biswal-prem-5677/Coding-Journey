*Below is a simple implementation of K-Means Clustering using Python's scikit-learn library:*

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# Generate synthetic dataset
X, _ = make_blobs(n_samples=400, centers=3, cluster_std=0.60, random_state=2)

X
```

```python
# Apply K-Means
kmeans = KMeans(n_clusters=5)
kmeans.fit(X)
y_kmeans = kmeans.predict(X)

# Plotting the clusters
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=50, cmap='viridis')
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.75)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('K-Means Clustering')
plt.show()
```

Explanation:

Data Generation: We create a synthetic dataset with 400 samples and 5 centers.

Model Training: The KMeans class from scikit-learn is used to fit the model and predict cluster assignments.

Visualization: Data points are plotted with colors representing cluster assignments. Red dots indicate the cluster centroids

Choosing the Optimal K:

To determine the appropriate number of clusters, the Elbow Method is commonly used. It involves plotting the WCSS against the number of clusters and identifying the "elbow point" where the rate of decrease sharply slows.

```python
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=0)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.title('Elbow Method for Optimal K')
plt.show()
```

In this plot, the optimal number of clusters is indicated by the point where the WCSS starts to level off, forming an "elbow".

```python
# Importing all Necessray Libraries
import numpy as np # for Statistical Computation
import matplotlib.pyplot as plt # for plotting graph
from sklearn.datasets import load_wine # dataset available in SKlearn
from sklearn.cluster import KMeans # Model KMEAN

```

```python
# Creating Custom Dataset with Make_blob and Plotting it
X,y = load_wine(return_X_y=True)

fig = plt.figure(0)
plt.grid(True)
plt.scatter(X[:,0],X[:,1])
plt.show()
```

```python
import seaborn as sns # Advance Library for Plotting
```

```python
#Find optimum number of cluster
sse = [] #SUM OF SQUARED ERROR (Empty list of Sum of Square Error )
for k in range(1,11):
	km = KMeans(n_clusters=k, random_state=2)
	km.fit(X)
	sse.append(km.inertia_)
```

```python
#Elbow Method : helps us to find the no of group to divide the data into a gop in any Unsupervised Learning

sns.set_style("whitegrid")
g=sns.lineplot(x=range(1,11), y=sse)

g.set(xlabel ="Number of cluster (k)",
	ylabel = "Sum Squared Error",
	title ='Elbow Method')

plt.show()
```
