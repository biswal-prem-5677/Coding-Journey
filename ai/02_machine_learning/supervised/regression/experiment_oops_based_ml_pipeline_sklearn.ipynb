```python
from abc import ABC, abstractmethod

# -------------------- ABSTRACTION --------------------
# Abstract base class defining a common ML interface
class MLModel(ABC):

    @abstractmethod
    def fit(self, X, y):
        pass

    @abstractmethod
    def predict(self, X):
        pass


# -------------------- INHERITANCE --------------------
# Concrete ML model inheriting from MLModel
class SimpleModel(MLModel):

    # -------------------- CONSTRUCTOR --------------------
    # Initializes model parameters (attributes)
    def __init__(self):
        self.mean = None

    # -------------------- ENCAPSULATION --------------------
    # Training logic hidden inside the class
    def fit(self, X, y):
        self.mean = sum(y) / len(y)

    # -------------------- POLYMORPHISM --------------------
    # Same method name (predict), different behavior possible for other models
    def predict(self, X):
        return [self.mean for _ in X]


# -------------------- OBJECT CREATION --------------------
# Creating object of the model class
model = SimpleModel()

# Sample data
X = [1, 2, 3, 4]
y = [10, 20, 30, 40]

# -------------------- METHOD CALLS --------------------
model.fit(X, y)
predictions = model.predict(X)

print("Predictions:", predictions)
```

```python
from abc import ABC, abstractmethod

# Abstraction: common ML interface
class Model(ABC):
    @abstractmethod
    def predict(self, X): pass

# Inheritance + Encapsulation
class MeanModel(Model):
    def __init__(self, y): self.mean = sum(y)/len(y)   # Constructor + Attribute
    def predict(self, X): return [self.mean]*len(X)   # Polymorphism

# Object creation & usage
X, y = [1,2,3], [10,20,30]
model = MeanModel(y)
print("Predictions:", model.predict(X))
```

```python
# -----------------------------------------------
# OBJECT-ORIENTED PROGRAMMING (OOPS) IN ML
# -----------------------------------------------

import pandas as pd
from abc import ABC, abstractmethod
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# ------------------------------------------------
# 1. CLASS & OBJECT
# DataLoader is a class that represents the concept
# of loading and managing datasets in ML.
# ------------------------------------------------
class DataLoader:
    def __init__(self, file_path):
        # Constructor initializes dataset
        # (Constructor & Attributes)
        self.data = pd.read_csv(file_path)

    def shape(self):
        # Encapsulation: internal data accessed via method
        return self.data.shape

    def summary(self):
        return self.data.describe()

    def select_features(self, columns):
        return self.data[columns]


# ------------------------------------------------
# 2. CONSTRUCTOR & ATTRIBUTES
# Preprocessor initializes preprocessing options
# using constructor arguments.
# ------------------------------------------------
class Preprocessor:
    def __init__(self, normalize=False, fill_missing=False):
        # Attributes store preprocessing configuration
        self.normalize = normalize
        self.fill_missing = fill_missing

    def process(self, data):
        # Encapsulation: preprocessing logic hidden inside class
        df = data.copy()

        if self.fill_missing:
            df = df.fillna(df.mean(numeric_only=True))

        if self.normalize:
            df = (df - df.min()) / (df.max() - df.min())

        return df


# ------------------------------------------------
# 3. ABSTRACTION
# MLModel is an abstract base class that defines
# a common interface for all ML models.
# ------------------------------------------------
class MLModel(ABC):

    @abstractmethod
    def fit(self, X, y):
        pass

    @abstractmethod
    def predict(self, X):
        pass

    @abstractmethod
    def evaluate(self, X, y):
        pass


# ------------------------------------------------
# 4. INHERITANCE + POLYMORPHISM
# LinearRegressionModel inherits from MLModel
# and provides its own implementation of methods.
# ------------------------------------------------
class LinearRegressionModel(MLModel):
    def __init__(self):
        # Constructor initializes ML algorithm
        self.model = LinearRegression()

    def fit(self, X, y):
        # Polymorphism: same method name, model-specific behavior
        self.model.fit(X, y)

    def predict(self, X):
        return self.model.predict(X)

    def evaluate(self, X, y):
        predictions = self.predict(X)
        return mean_squared_error(y, predictions)


# ------------------------------------------------
# 5. END-TO-END ML PIPELINE (OBJECT INTERACTION)
# ------------------------------------------------

# Create object of DataLoader
loader = DataLoader("/content/sample_data/california_housing_train.csv")

# Load and inspect data
data = loader.data
print("Dataset Shape:", loader.shape())

# Create object of Preprocessor
preprocessor = Preprocessor(normalize=True, fill_missing=True)

# Preprocess data
data = preprocessor.process(data)

# Split features and target
# Assuming 'median_house_value' is the target column in the California Housing dataset
X = data.drop("median_house_value", axis=1)
y = data["median_house_value"]

# Create object of ML model (Polymorphism in action)
model = LinearRegressionModel()

# Train model
model.fit(X, y)

# Evaluate model
error = model.evaluate(X, y)
print("Model Error:", error)

# Predict
predictions = model.predict(X.head())
print("Predictions:", predictions)
```
