# **Linera Regression Implementation **

1.Pizza price prediction

X = Pizza Diameter     
y = Price

Predict for X = 8,13,14

Mean

```python
import numpy as np

X = np.array([1,3,5])
y = np.array([30000,45000,60000])
meanX = np.mean(X)
meany = np.mean(y)
```

Deviation

```python
DeviationX = X - meanX
DeviationY = y - meany
```

Slope

```python
sum_product_deviations = np.sum(DeviationX * DeviationY)
sum_squared_deviationX = np.sum(DeviationX**2)
M = sum_product_deviations / sum_squared_deviationX
print(f'{sum_product_deviations=}')
print(f'{sum_squared_deviationX=}')
print(f'{M=}')
```

Intercept

```python
C = meany - M * meanX
print(f'{C=}')
```

Method Defining For Prediction

```python
def predict_price(X):

  return  M * X + C
```

Method call for Prediction


```python
X_predict = np.array([4])
predicted_prices = predict_price(X_predict)
print("Predicted prices for X =", X_predict, ":", predicted_prices)
```

2.Marks Prediction Based On Study Hour

X = Study Hour       
y = Score

Predict for X = 5hrs

```python
X = [2, 4, 6]
Y = [50, 65, 80]

mean_x = sum(X)/len(X)
mean_y = sum(Y)/len(Y)

num = sum((x-mean_x)*(y-mean_y) for x,y in zip(X,Y))
den = sum((x-mean_x)**2 for x in X)
m = num/den
b = mean_y - m*mean_x

X_pred = 5
Y_pred = m*X_pred + b

print(f"Equation: Y = {m:.4f}*X + {b:.4f}")
print(f"Predicted score for {X_pred} hours = {Y_pred:.2f}")
```
