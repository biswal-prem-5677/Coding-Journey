```python
# ============================================
# LEVEL 1: DISCRIMINATIVE MODEL
# ============================================

import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

X = np.random.randn(1000, 2)
y = (X[:,0] + X[:,1] > 0).astype(int)

X_train, X_test, y_train, y_test = train_test_split(X, y)

model_disc = LogisticRegression()
model_disc.fit(X_train, y_train)

probs = model_disc.predict_proba(X_test)

print("P(y|x):\n", probs[:5])
```

```python
# ============================================
# LEVEL 2: GENERATIVE MODEL
# ============================================

from sklearn.naive_bayes import GaussianNB

model_gen = GaussianNB()
model_gen.fit(X_train, y_train)

preds = model_gen.predict(X_test)

print("Predictions:", preds[:10])
```

```python
from sklearn.metrics import accuracy_score

acc_disc = accuracy_score(y_test, model_disc.predict(X_test))
acc_gen = accuracy_score(y_test, model_gen.predict(X_test))

print("Discriminative Accuracy:", acc_disc)
print("Generative Accuracy:", acc_gen)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X_test[:,0], X_test[:,1], c=model_disc.predict(X_test), cmap='coolwarm')
plt.title("Discriminative Boundary")
plt.show()

plt.scatter(X_test[:,0], X_test[:,1], c=model_gen.predict(X_test), cmap='coolwarm')
plt.title("Generative Boundary")
plt.show()
```
