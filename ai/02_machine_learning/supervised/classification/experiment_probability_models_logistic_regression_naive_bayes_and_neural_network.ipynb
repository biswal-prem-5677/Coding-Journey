```python
# ============================================
# LOGISTIC REGRESSION AS PROBABILITY MODEL
# ============================================

import numpy as np
from sklearn.linear_model import LogisticRegression

X = np.random.randn(1000, 2)
y = (X[:,0] + X[:,1] > 0).astype(int)

model = LogisticRegression()
model.fit(X, y)

probs = model.predict_proba(X)

print("Probabilities:", probs[:5])
```

```python
from sklearn.naive_bayes import GaussianNB

model_nb = GaussianNB()
model_nb.fit(X, y)

probs_nb = model_nb.predict_proba(X)

print("NB Probabilities:", probs_nb[:5])
```

```python
from sklearn.linear_model import LinearRegression

X = np.random.randn(1000,1)
y = 3*X.squeeze() + np.random.randn(1000)*0.5

model_lr = LinearRegression()
model_lr.fit(X, y)

preds = model_lr.predict(X)

print("Predictions:", preds[:5])
```

```python
import torch
import torch.nn as nn

model_nn = nn.Sequential(
    nn.Linear(2, 10),
    nn.ReLU(),
    nn.Linear(10, 2),
    nn.Softmax(dim=1)
)

x = torch.randn(5,2)

output = model_nn(x)

print("NN Probabilities:\n", output)
```
