```python
class MDP:
    def __init__(self, states, actions, transitions, gamma=0.9):
        self.states = states
        self.actions = actions
        self.transitions = transitions
        self.gamma = gamma
        self.V = {s: 0 for s in states}

    def value_iteration(self, iterations=50):
        for _ in range(iterations):
            new_V = {}
            for s in self.states:
                values = []
                for a in self.actions:
                    next_state, reward = self.transitions[s][a]
                    values.append(reward + self.gamma * self.V[next_state])
                new_V[s] = max(values)
            self.V = new_V
        return self.V


if __name__ == "__main__":
    states = ['A', 'B']
    actions = ['left', 'right']

    transitions = {
        'A': {'left': ('A', 0), 'right': ('B', 1)},
        'B': {'left': ('A', 0), 'right': ('B', 2)}
    }

    mdp = MDP(states, actions, transitions)
    print(mdp.value_iteration())
```
