```python
import numpy as np

states = ['A', 'B']
actions = ['left', 'right']

P = {
    'A': {'left': ('A', 0), 'right': ('B', 1)},
    'B': {'left': ('A', 0), 'right': ('B', 2)}
}

gamma = 0.9

# Initial random policy
policy = {s: np.random.choice(actions) for s in states}
V = {s: 0 for s in states}

def policy_evaluation(policy, V, iterations=10):
    for _ in range(iterations):
        new_V = {}
        for s in states:
            a = policy[s]
            next_state, reward = P[s][a]
            new_V[s] = reward + gamma * V[next_state]
        V = new_V
    return V

def policy_improvement(V):
    new_policy = {}
    for s in states:
        values = {}
        for a in actions:
            next_state, reward = P[s][a]
            values[a] = reward + gamma * V[next_state]
        new_policy[s] = max(values, key=values.get)
    return new_policy

for i in range(5):
    V = policy_evaluation(policy, V)
    policy = policy_improvement(V)

print("Optimal Policy:", policy)
print("Value Function:", V)
```
