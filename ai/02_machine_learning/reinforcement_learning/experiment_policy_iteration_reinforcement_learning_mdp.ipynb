```python
class PolicyIteration:
    def __init__(self, states, actions, transitions, gamma=0.9):
        self.states = states
        self.actions = actions
        self.transitions = transitions
        self.gamma = gamma
        self.V = {s: 0 for s in states}
        self.policy = {s: actions[0] for s in states}

    def policy_evaluation(self, iterations=20):
        for _ in range(iterations):
            new_V = {}
            for s in self.states:
                a = self.policy[s]
                next_state, reward = self.transitions[s][a]
                new_V[s] = reward + self.gamma * self.V[next_state]
            self.V = new_V

    def policy_improvement(self):
        stable = True
        for s in self.states:
            old_action = self.policy[s]
            values = {}

            for a in self.actions:
                next_state, reward = self.transitions[s][a]
                values[a] = reward + self.gamma * self.V[next_state]

            best_action = max(values, key=values.get)
            self.policy[s] = best_action

            if old_action != best_action:
                stable = False

        return stable

    def run(self):
        while True:
            self.policy_evaluation()
            if self.policy_improvement():
                break
        return self.policy, self.V


if __name__ == "__main__":
    states = ['A', 'B']
    actions = ['left', 'right']

    transitions = {
        'A': {'left': ('A', 0), 'right': ('B', 1)},
        'B': {'left': ('A', 0), 'right': ('B', 2)}
    }

    solver = PolicyIteration(states, actions, transitions)
    policy, values = solver.run()

    print("Optimal Policy:", policy)
    print("Optimal Values:", values)
```
