```python
import random

class QLearning:
    def __init__(self, states, actions, transitions,
                 alpha=0.1, gamma=0.9, epsilon=0.1):
        self.states = states
        self.actions = actions
        self.transitions = transitions
        self.alpha = alpha
        self.gamma = gamma
        self.epsilon = epsilon

        self.Q = {(s, a): 0 for s in states for a in actions}

    def choose_action(self, state):
        if random.uniform(0, 1) < self.epsilon:
            return random.choice(self.actions)
        return max(self.actions, key=lambda a: self.Q[(state, a)])

    def train(self, episodes=100):
        for _ in range(episodes):
            state = random.choice(self.states)

            for _ in range(10):
                action = self.choose_action(state)
                next_state, reward = self.transitions[state][action]

                best_next = max(self.Q[(next_state, a)] for a in self.actions)

                self.Q[(state, action)] += self.alpha * (
                    reward + self.gamma * best_next - self.Q[(state, action)]
                )

                state = next_state

        return self.Q


if __name__ == "__main__":
    states = ['A', 'B']
    actions = ['left', 'right']

    transitions = {
        'A': {'left': ('A', 0), 'right': ('B', 1)},
        'B': {'left': ('A', 0), 'right': ('B', 2)}
    }

    agent = QLearning(states, actions, transitions)
    Q = agent.train()

    print("Learned Q-values:")
    for k, v in Q.items():
        print(k, round(v, 2))
```
