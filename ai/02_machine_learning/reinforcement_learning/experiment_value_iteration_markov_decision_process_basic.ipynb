```python
import numpy as np

# States
states = ['A', 'B']

# Actions
actions = ['left', 'right']

# Transition Probabilities
P = {
    'A': {'left': ('A', 0), 'right': ('B', 1)},
    'B': {'left': ('A', 0), 'right': ('B', 2)}
}

gamma = 0.9

# Initialize value function
V = {s: 0 for s in states}

def value_iteration(V, iterations=10):
    for _ in range(iterations):
        new_V = {}
        for s in states:
            values = []
            for a in actions:
                next_state, reward = P[s][a]
                values.append(reward + gamma * V[next_state])
            new_V[s] = max(values)
        V = new_V
    return V

V = value_iteration(V)
print("Optimal Value Function:", V)
```
