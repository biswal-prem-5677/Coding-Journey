## Q4) Launch the interface and understand the various UI components.

```python
import gradio as gr


def sentence_builder(quantity, animal, countries, place, activity_list, morning):
    return f"""The {quantity} {animal}s from {" and ".join(countries)} went to the {place} where they {" and ".join(activity_list)} until the {"morning" if morning else "night"}"""


demo = gr.Interface(
    fn = sentence_builder,
    inputs = [
        gr.Slider(0, 100, value=2, label="Count", info="Choose between 0 and 100"),
        gr.Dropdown(
            ["cat", "dog", "bird","Fish"], label="Animal", info="Will add more animals later!"
        ),
        gr.CheckboxGroup(["USA", "Japan","India", "Pakistan"], label="Countries", info="Where are they from?"),
        gr.Radio(["park", "zoo", "road"], label="Location", info="Where did they go?"),
        gr.Dropdown(
            ["ran", "swam", "ate", "slept"], value=["swam", "slept"], multiselect=True, label="Activity", info="PLease Select your favorite Activity"
        ),
        gr.Checkbox(label="Morning", info="Did they do it in the morning?"),
    ],
    outputs = "text",
    examples=[
        [2, "cat", ["Japan", "Pakistan"], "park", ["ate", "swam"], True],
        [4, "dog", ["Japan"], "zoo", ["ate", "swam"], False],
        [10, "bird", ["USA", "Pakistan"], "road", ["ran"], False],
        [8, "cat", ["Pakistan"], "zoo", ["ate"], True],
    ]
)
demo.launch()
```

Q5) Using an image filter

```python
import numpy as np
import gradio as gr

def sepia(input_img):
    sepia_filter = np.array([
        [0.393, 0.769, 0.189],
        [0.349, 0.686, 0.168],
        [0.272, 0.534, 0.131]
    ])
    sepia_img = input_img.dot(sepia_filter.T)
    sepia_img /= sepia_img.max()
    return sepia_img

demo = gr.Interface(sepia, gr.Image(), "image")
demo.launch()
```

```python
from transformers import pipeline
import gradio as gr

# Initialize the chatbot by explicitly specifying the task
chatbot = pipeline('conversational', model="facebook/blenderbot-3B")

# Define the chatbot function
def vanilla_chatbot(message, history):
    # The history parameter in gr.ChatInterface handles conversation history automatically.
    # We'll use the current message to get a response.
    response = chatbot(message)[0]['generated_text']
    return response

# Create a Gradio interface
demo_chatbot = gr.ChatInterface(
    vanilla_chatbot,
    title="Mashdemy Chatbot",
    description="A simple chatbot interface"
)

# Launch the demo
demo_chatbot.launch(share=True)
```

```python
import gradio as gr
import numpy as np
from scipy.io.wavfile import write

def generate_music(prompt):
    # This is a dummy function. In a real application, you would use a music generation model here.
    # For demonstration, we'll create a simple sine wave.
    sample_rate = 44100  # samples per second
    duration = 5.0  # seconds
    frequency = 440.0  # Hz (A4 note)

    t = np.linspace(0., duration, int(sample_rate * duration), endpoint=False)
    amplitude = np.iinfo(np.int16).max * 0.5  # Max amplitude for 16-bit PCM
    data = amplitude * np.sin(2 * np.pi * frequency * t)
    audio_data = data.astype(np.int16)

    output_filename = "dummy_music.wav"
    write(output_filename, sample_rate, audio_data)

    return output_filename

inf = gr.Interface(generate_music,title = "Mashdemy Demo Music Generator App", description = "Type in the kind of music you prefer and click submit", inputs =["text"], outputs=["audio"], examples = ["lo-fi music with a soothing melody", "pop music"])
inf.launch(share = True)
```

```python
from transformers import pipeline
import gradio as gr

# Initialize the chatbot with a better model, explicitly specifying the task
chatbot = pipeline('conversational', model="facebook/blenderbot-3B")

# Initialize conversation history
history = []

# Define the chatbot function
def vanilla_chatbot(message, history):
    # Append the latest user message to the history
    history.append(message)

    # Join the history into a single string for the model
    conversation = " ".join(history)

    # Generate a response using the chatbot
    response = chatbot(conversation)['generated_text']

    # Append the response to history
    history.append(response)

    # Return the response and updated history
    return response, history

# Create a Gradio interface
demo_chatbot = gr.ChatInterface(
    vanilla_chatbot,
    title="Mashdemy Chatbot",
    description="A simple chatbot interface"
)

# Launch the demo
demo_chatbot.launch()
```
