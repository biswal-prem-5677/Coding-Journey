```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as tk
import numpy as np

df=pd.read_excel(r"/content/sales analysis sheet final.xlsx",usecols=[1,2,3,4,5],)
df.head(10)


frenchise_owner=df.iloc[0:10]['Frenchise Owner']
net_qty_supplied=df.iloc[0:10]['Net Qty Supplied(pieces)']
net_qty_sold_online=df.iloc[0:10]['Net Qty Sold Online(pieces)']
net_qty_sold_offline=df.iloc[0:10]['Net Qty Sold Offline(pieces)']

plt.figure(figsize=(15,15))
barwidth=0.3


x=np.arange(len(frenchise_owner))

# Define a function to add bar values

def add_bar_values(bars):
    offset = 5                   # Adjust vertical offset for text
    for bar in bars:
        height = bar.get_height()
        x = bar.get_x() + bar.get_width() / 2
        y = height + offset                      # Adjust vertical position

        plt.text(x, y, f'{int(height)}', ha='center', va='bottom',
                     fontsize=12, color='black')

bar1=plt.bar(x-barwidth,net_qty_supplied,width=barwidth,color='red',label='Net Qty Supplied')
bar2=plt.bar(x,net_qty_sold_online,width=barwidth,color='purple', label='Net Qty Sold Online')
bar3=plt.bar(x+barwidth,net_qty_sold_offline,width=barwidth,color='pink',label='Net Qty Sold Offline')




add_bar_values(bar1)
add_bar_values(bar2)
add_bar_values(bar3)

plt.xticks(x,frenchise_owner,rotation=45)
plt.subplots_adjust(bottom=0.4)


plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df=pd.read_excel(r"/content/Sales Analysis Sheet.xlsx",usecols=[1,2,3,4],)
df.head()


frenchise_owner=df.iloc[9:19]['Frenchise Owner']
net_qty_supplied=df.iloc[9:19]['Net Qty Supplied(pieces)']
net_qty_sold_online=df.iloc[9:19]['Net Qty Sold Online(pieces)']
net_qty_sold_offline=df.iloc[9:19]['Net Qty Sold Offline(pieces)']


plt.figure(figsize=(25,16))
barwidth=0.3


x=np.arange(len(frenchise_owner))

# Define a function to add bar values

def add_bar_values(bars):
    offset = 5                   # Adjust vertical offset for text
    for bar in bars:
        height = bar.get_height()
        x = bar.get_x() + bar.get_width() / 2
        y = height/2                       # Adjust vertical position

        plt.text(x, y, str(int(height)), ha='center', va='',
                     fontsize=12, color='black')

bar1=plt.bar(x-barwidth, net_qty_supplied, width=barwidth, label='Net Qty Supplied', color='blue')
bar2=plt.bar(x, net_qty_sold_online, width=barwidth, label='Net Qty Sold Online', color='dodgerblue')
bar3=plt.bar(x+barwidth, net_qty_sold_offline, width=barwidth, label='Net Qty Sold Offline', color='aqua')


add_bar_values(bar1)
add_bar_values(bar2)
add_bar_values(bar3)

plt.title("-: Louis Vuitton Sales Overview :-",fontsize=20)
plt.xlabel("Frenchise Owner →",fontsize=14)
plt.ylabel("Quantity In Pieces →",fontsize=14)

plt.xticks(x,frenchise_owner,rotation=45,fontsize=14)
plt.subplots_adjust(bottom=0.4)

plt.legend()
plt.show()
```

```python
 import numpy as np
 import matplotlib.pyplot as plt
 people = ('A','B','C','D','E','F','G','H')
 segments = 4
 # generate some multi-dimensional data & arbitrary labels
 data = 3 + 10* np.random.rand(segments, len(people))
 percentages = (np.random.randint(5,20, (len(people), segments)))
 y_pos = np.arange(len(people))
 fig = plt.figure(figsize=(10,8))
 ax = fig.add_subplot(111)
 colors ='rgbwmc'
 patch_handles = []
 left = np.zeros(len(people)) # left alignment of data starts at zero
 for i, d in enumerate(data):
     patch_handles.append(ax.barh(y_pos, d,
color=colors[i%len(colors)], align='center',
left=left))
 # accumulate the left-hand offsets
 left += d
 # go through all of the bar segments and annotate
 for j in range(len(patch_handles)):
     for i, patch in enumerate(patch_handles[j].get_children()):
         bl = patch.get_xy()
 x = 0.5*patch.get_width() + bl[0]
 y = 0.5*patch.get_height() + bl[1]
 ax.text(x,y, "%d%%" % (percentages[i,j]), ha='center')
 ax.set_yticks(y_pos)
 ax.set_yticklabels(people)
 ax.set_xlabel('Distance')
 plt.show()
```
