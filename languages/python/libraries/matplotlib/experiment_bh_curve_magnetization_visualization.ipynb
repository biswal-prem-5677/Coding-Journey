```python
import matplotlib.pyplot as plt
import numpy as np

# B-H Curve Plot
H = np.linspace(0, 1200, 1000)  # Magnetic field strength (H) in A/m
B_s = 1.5  # Saturation flux density (Tesla)
B = B_s * (1 - np.exp(-H / 200))  # Simulated B-H curve

# Points for Remanence and Coercivity
Hc = 500  # Coercive force in A/m
Br = 1.2  # Remanent flux density in Tesla

# Create the plot
plt.figure(figsize=(8, 6))
plt.plot(H, B, label="B-H Curve", color='b', linewidth=2)

# Highlight key points (Saturation, Remanence, Coercivity)
plt.scatter([Hc], [Br], color='r', zorder=5)
plt.text(500, 1.15, "Br (Remanence)", horizontalalignment='left', verticalalignment='bottom', color='r')
plt.text(200, 1.4, "Saturation", horizontalalignment='left', verticalalignment='bottom', color='b')
plt.text(700, 0.5, "Hc (Coercivity)", horizontalalignment='right', verticalalignment='top', color='r')

# Labeling the axes
plt.xlabel('Magnetic Field Strength (H) [A/m]', fontsize=12)
plt.ylabel('Magnetic Flux Density (B) [Tesla]', fontsize=12)
plt.title('B-H Curve (Magnetization Curve)', fontsize=14)

# Add grid and legend
plt.grid(True)
plt.axhline(0, color='black',linewidth=0.5)
plt.axvline(0, color='black',linewidth=0.5)
plt.legend()

# Show the plot
plt.show()
```
