# **List Comprehension**

```python
#Example: Clean Missing Data
data = [10, None, 25, None, 40]

cleaned = [x for x in data if x is not None]
print(cleaned)
```

```python
#Normalize Data
data = [10, 20, 30, 40]

normalized = [x / max(data) for x in data]
print(normalized)
```

# **Dictionary Comprehension**

```python
#Example: Feature Scaling
data = {'a': 10, 'b': 20, 'c': 30}

scaled = {k: v / 10 for k, v in data.items()}
print(scaled)
```

# **Set Comprehension (Remove Duplicates)**

```python
data = [1, 2, 2, 3, 4, 4]

unique = {x for x in data}
print(unique)
```

# **Using map() (Transformation)**

```python
#Example: Convert Strings to Integers
data = ['1', '2', '3']

nums = list(map(int, data))
print(nums)
```

```python
#Example with Lambda
data = [1, 2, 3]

squared = list(map(lambda x: x**2, data))
print(squared)
```

# **Using filter() (Selection)**

```python
#Example: Keep Even Numbers
data = [1, 2, 3, 4, 5, 6]

even = list(filter(lambda x: x % 2 == 0, data))
print(even)
```

# **Using reduce() (Aggregation)**

```python
from functools import reduce

data = [1, 2, 3, 4]

sum_all = reduce(lambda x, y: x + y, data)
print(sum_all)
```

# **Combined Pipeline (Real ML Preprocessing)**

```python
#Example: Clean → Transform → Filter

data = [10, None, 25, 30, None, 50]

result = list(
    map(
        lambda x: x / 10,
        filter(
            lambda x: x is not None and x > 20,
            data
        )
    )
)

print(result)
```

# **Nested Comprehension (Advanced)**

```python
#Flatten Data
matrix = [[1, 2], [3, 4], [5, 6]]

flat = [item for row in matrix for item in row]
print(flat)
```

# **Conditional Transformation**

```python
data = [10, 15, 20, 25]

processed = [x if x % 2 == 0 else 0 for x in data]
print(processed)
```

# **Mini Example**

```python
# Raw dataset
data = ["10", "20", "NA", "30", "40"]

processed = [
    int(x) / 10
    for x in data
    if x != "NA"
]

print(processed)
```
