```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris

# Load dataset
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Add target
df["target"] = iris.target

# Add missing values (demo)
df.iloc[0, 0] = np.nana

# Fill missing values
df.fillna(df.mean(), inplace=True)

# Remove duplicates
df.drop_duplicates(inplace=True)

# Rename columns
df.rename(columns={
    "sepal length (cm)": "sepal_length",
    "sepal width (cm)": "sepal_width",
    "petal length (cm)": "petal_length",
    "petal width (cm)": "petal_width"
}, inplace=True)

# Normalize
df["sepal_length"] = (
    (df["sepal_length"] - df["sepal_length"].min())
    /
    (df["sepal_length"].max() -
     df["sepal_length"].min())
)

# Create feature
df["total_length"] = (
    df["sepal_length"] +
    df["petal_length"]
)

# Save processed data
df.to_csv("processed_iris.csv", index=False)

print(df.head())
```
