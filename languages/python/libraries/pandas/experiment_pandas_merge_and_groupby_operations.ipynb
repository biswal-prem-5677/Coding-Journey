```python
import pandas as pd

# Employee Data

employees = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104],
    "Name": ["Alice", "Bob", "Charlie", "David"]
})

# Department Data

departments = pd.DataFrame({
    "Emp_ID": [101, 102, 104, 105],
    "Department": ["HR", "IT", "Finance", "Marketing"]
})

print("Employees")
print(employees)

print("\nDepartments")
print(departments)
```

```python
inner_merge = pd.merge(
    employees,
    departments,
    on="Emp_ID",
    how="inner"
)

print(inner_merge)
```

```python
left_merge = pd.merge(
    employees,
    departments,
    on="Emp_ID",
    how="left"
)

print(left_merge)
```

```python
right_merge = pd.merge(
    employees,
    departments,
    on="Emp_ID",
    how="right"
)

print(right_merge)
```

```python
outer_merge = pd.merge(
    employees,
    departments,
    on="Emp_ID",
    how="outer"
)

print(outer_merge)
```

```python
sales = pd.DataFrame({
    "Department": [
        "IT",
        "IT",
        "HR",
        "HR",
        "Finance",
        "Finance"
    ],

    "Employee": [
        "A",
        "B",
        "C",
        "D",
        "E",
        "F"
    ],

    "Salary": [
        50000,
        60000,
        45000,
        55000,
        70000,
        80000
    ]
})

print(sales)
```

```python
grouped = sales.groupby("Department")

print(grouped["Salary"].sum())
```

```python
summary = sales.groupby("Department")["Salary"].agg(
    ["count", "sum", "mean", "max", "min"]
)

print(summary)
```

```python
import pandas as pd

# ============================================================
# EMPLOYEE DATA
# ============================================================

employees = pd.DataFrame({
    "Emp_ID": [101,102,103,104,105],
    "Name": ["Alice","Bob","Charlie","David","Emma"],
    "Dept_ID": [1,2,1,3,2]
})

# ============================================================
# DEPARTMENT DATA
# ============================================================

departments = pd.DataFrame({
    "Dept_ID": [1,2,3],
    "Department": ["HR","IT","Finance"]
})

# ============================================================
# SALARY DATA
# ============================================================

salary = pd.DataFrame({
    "Emp_ID": [101,102,103,104,105],
    "Salary": [50000,70000,60000,90000,75000]
})

# ============================================================
# MERGING DATAFRAMES
# ============================================================

merged = pd.merge(
    employees,
    departments,
    on="Dept_ID"
)

merged = pd.merge(
    merged,
    salary,
    on="Emp_ID"
)

print("\nMerged DataFrame")
print(merged)

# ============================================================
# GROUPING
# ============================================================

department_summary = merged.groupby(
    "Department"
)["Salary"].agg(
    [
        "count",
        "sum",
        "mean",
        "max",
        "min"
    ]
)

print("\nDepartment Summary")
print(department_summary)
```
