# **Basic Matrix Operations in NumPy**

(a) Matrix Creation

```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

print(A)
print(B)
```

(b) Addition

```python
C = A + B
print(C)
```

(c) Subtraction

```python
C = A - B
print(C)
```

(d) Element-wise Multiplication

```python
C = A * B
print(C)
```

(e) Matrix Multiplication

```python
C = A @ B
# OR
C = np.dot(A, B)

print(C)
```

(f) Transpose

```python
print(A.T)
```

(g) Inverse

```python
A_inv = np.linalg.inv(A)
print(A_inv)
```

# **Broadcasting in NumPy**

Scalar + Matrix

```python
A = np.array([[1, 2],
              [3, 4]])

B = A + 10
print(B)
```

Vector + Matrix (Row-wise)

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([10, 20, 30])

C = A + B
print(C)
```

Column Broadcasting

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[10],
              [20]])

C = A + B
print(C)
```

# **Real ML Use Case**

```python
X = np.array([[1, 2, 3],
              [4, 5, 6]])

mean = np.mean(X, axis=0)
X_norm = X - mean   # Broadcasting

print(X_norm)
```
