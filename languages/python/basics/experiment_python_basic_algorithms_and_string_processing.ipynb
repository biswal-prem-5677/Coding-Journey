# **1️⃣ Prime Number Generator**

```python
def primes_upto(n):
    primes = []

    for num in range(2, n + 1):
        is_prime = True

        for p in primes:
            if p * p > num:
                break
            if num % p == 0:
                is_prime = False
                break

        if is_prime:
            primes.append(num)

    return primes

print(primes_upto(50))
```

# **2️⃣ Longest Word in a Sentence**

```python
def longest_word(sentence):
    words = sentence.split()

    longest = ""

    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest

text = input("Enter sentence: ")
print("Longest word:", longest_word(text))
```

# **3️⃣ Count Character Frequency**

```python
def char_frequency(text):
    freq = {}

    for ch in text:
        if ch in freq:
            freq[ch] += 1
        else:
            freq[ch] = 1

    return freq

print(char_frequency("programming"))
```

# **4️⃣ Palindrome Checker (Ignoring Spaces)**

```python
def is_palindrome(s):
    s = s.replace(" ", "").lower()

    left = 0
    right = len(s) - 1

    while left < right:
        if s[left] != s[right]:
            return False

        left += 1
        right -= 1

    return True


text = input("Enter text: ")
print("Palindrome:", is_palindrome(text))
```

# **5️⃣ Simple Task Scheduler (Priority Logic)**

```python
tasks = []

def add_task(name, priority):
    tasks.append((priority, name))
    tasks.sort()

def show_tasks():
    for p, n in tasks:
        print(f"Priority {p} -> {n}")

add_task("Study AI", 2)
add_task("Do assignment", 1)
add_task("Workout", 3)

show_tasks()
```
