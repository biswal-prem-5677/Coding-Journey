```python
def custom_filter(func, iterable):
    result = []

    for item in iterable:
        if func(item):
            result.append(item)

    return result

print(custom_filter(lambda x: x % 2 == 0, [1,2,3,4,5,6]))
```

```python
def count_calls(func):

    count = 0

    def wrapper(*args, **kwargs):
        nonlocal count

        count += 1

        print(f"Called {count} times")

        return func(*args, **kwargs)

    return wrapper

@count_calls
def greet():
    print("Hello")

greet()
greet()
greet()
```

```python
import time

class Timer:

    def __enter__(self):
        self.start = time.time()

    def __exit__(self, exc_type, exc_val, exc_tb):
        print(
            f"Elapsed: {time.time() - self.start:.4f} sec"
        )

with Timer():
    sum(range(10_000_000))
```

```python
def find_key(data, target):

    if target in data:
        return data[target]

    for value in data.values():
        if isinstance(value, dict):
            result = find_key(value, target)

            if result:
                return result

    return None

nested = {
    "a": {
        "b": {
            "c": 100
        }
    }
}

print(find_key(nested, "c"))
```

```python
def pipeline(data, functions):

    for func in functions:
        data = func(data)

    return data

functions = [
    lambda x: x + 5,
    lambda x: x * 2,
    lambda x: x - 3
]

print(pipeline(10, functions))
```
