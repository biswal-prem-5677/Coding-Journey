```python
def deep_merge(dict1, dict2):

    result = dict1.copy()

    for key, value in dict2.items():

        if (
            key in result and
            isinstance(result[key], dict) and
            isinstance(value, dict)
        ):
            result[key] = deep_merge(
                result[key],
                value
            )
        else:
            result[key] = value

    return result


a = {
    "database": {
        "host": "localhost"
    }
}

b = {
    "database": {
        "port": 5432
    }
}

print(deep_merge(a, b))
```

```python
def find_difference(old, new):

    diff = {}

    for key in new:

        if key not in old:
            diff[key] = "Added"

        elif old[key] != new[key]:
            diff[key] = (
                old[key],
                new[key]
            )

    return diff


before = {
    "name": "John",
    "age": 20
}

after = {
    "name": "John",
    "age": 21
}

print(find_difference(before, after))
```

```python
def max_depth(data):

    if not isinstance(data, dict):
        return 0

    if not data:
        return 1

    return 1 + max(
        max_depth(value)
        for value in data.values()
    )


sample = {
    "a": {
        "b": {
            "c": {
                "d": 1
            }
        }
    }
}

print(max_depth(sample))
```

```python
def format_bytes(size):

    units = [
        "B",
        "KB",
        "MB",
        "GB",
        "TB"
    ]

    for unit in units:

        if size < 1024:
            return f"{size:.2f} {unit}"

        size /= 1024

print(format_bytes(123456789))
```

```python
def similarity(text1, text2):

    set1 = set(text1.lower().split())
    set2 = set(text2.lower().split())

    intersection = len(set1 & set2)
    union = len(set1 | set2)

    return intersection / union


print(
    similarity(
        "AI will change the world",
        "AI is changing the world"
    )
)
```
