```python
def create_command(action):

    def execute():
        print(f"Executing {action}")

    return execute

save = create_command("Save")
delete = create_command("Delete")

save()
delete()
```

```python
import time

def requests_per_second():

    timestamps = []

    def record():

        timestamps.append(time.time())

        current = time.time()

        timestamps[:] = [
            t
            for t in timestamps
            if current - t < 1
        ]

        return len(timestamps)

    return record
```

```python
def my_reduce(func, iterable):

    result = iterable[0]

    for item in iterable[1:]:
        result = func(result, item)

    return result

print(
    my_reduce(
        lambda x, y: x + y,
        [1,2,3,4]
    )
)
```

```python
def traverse(tree):

    print(tree["value"])

    for child in tree.get("children", []):
        traverse(child)
```

```python
import heapq

tasks = []

def schedule(priority, task):

    heapq.heappush(
        tasks,
        (priority, task)
    )

def run():

    while tasks:

        _, task = heapq.heappop(tasks)

        task()

schedule(2, lambda: print("Medium"))
schedule(1, lambda: print("High"))
schedule(3, lambda: print("Low"))

run()
```
