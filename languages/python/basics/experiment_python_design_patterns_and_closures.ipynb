```python
def singleton(cls):

    instance = None

    def get_instance():

        nonlocal instance

        if instance is None:
            instance = cls()

        return instance

    return get_instance

@singleton
class Database:
    pass

db1 = Database()
db2 = Database()

print(db1 is db2)
```

```python
def serialize(obj):

    return {
        key: value
        for key, value in obj.__dict__.items()
    }

class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

s = Student("Priyabrata", 20)

print(serialize(s))
```

```python
def execute_queue(tasks):

    results = []

    for task in tasks:
        results.append(task())

    return results

tasks = [
    lambda: 10 + 20,
    lambda: 5 * 6,
    lambda: 100 / 4
]

print(execute_queue(tasks))
```

```python
def undo_manager():

    history = []

    def add(action):
        history.append(action)

    def undo():
        return history.pop()

    return add, undo

add, undo = undo_manager()

add("Write code")
add("Run code")

print(undo())
```

```python
def get_nested(data, path):

    keys = path.split(".")

    for key in keys:
        data = data[key]

    return data

data = {
    "user": {
        "profile": {
            "name": "Priyabrata"
        }
    }
}

print(
    get_nested(
        data,
        "user.profile.name"
    )
)
```
