```python
import time

def create_token(ttl):

    created = time.time()

    def is_valid():

        return (
            time.time()
            - created
        ) < ttl

    return is_valid


token = create_token(10)

print(token())
```

```python
def search(
    query,
    documents
):

    query = query.lower()

    return [

        doc

        for doc in documents

        if query in doc.lower()
    ]


docs = [
    "AI is amazing",
    "Python programming",
    "Machine Learning"
]

print(search("python", docs))
```

```python
def frequency_map(text):

    result = {}

    for char in text:

        result[char] = (
            result.get(char, 0)
            + 1
        )

    return result


print(
    frequency_map(
        "MISSISSIPPI"
    )
)
```

```python
def rotate_matrix(matrix):

    return [
        list(row)

        for row

        in zip(*matrix[::-1])
    ]


matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

print(
    rotate_matrix(matrix)
)
```

```python
def parse_query(url):

    query = url.split("?")[1]

    pairs = query.split("&")

    result = {}

    for pair in pairs:

        key, value = pair.split("=")

        result[key] = value

    return result


print(
    parse_query(
        "site.com?id=5&name=john"
    )
)
```

```python
def evaluate_rules(
    data,
    rules
):

    for rule in rules:

        if not rule(data):
            return False

    return True


rules = [
    lambda x: x["age"] > 18,
    lambda x: x["salary"] > 30000
]

user = {
    "age": 25,
    "salary": 50000
}

print(
    evaluate_rules(
        user,
        rules
    )
)
```
