## Irritating Function Fun

```python
import time

def irritate_someone(message="Are we there yet?", count=5, delay=0.5):
  """
  Prints an irritating message multiple times with a delay.

  Args:
    message (str): The message to print.
    count (int): How many times to print the message.
    delay (float): Delay in seconds between messages.
  """
  print("Starting the irritation...")
  for i in range(count):
    print(message)
    time.sleep(delay)
  print("Irritation complete (for now)!\n")
```

### Demonstration

```python
# Try running this cell!
irritate_someone()

# Or with a custom message and more repetitions:
irritate_someone(message="Tap, tap, tap...", count=3, delay=1)
```
