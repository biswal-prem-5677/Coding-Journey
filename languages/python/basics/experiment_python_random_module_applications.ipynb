```python
import random

def roll_dice(sides=6):
    return random.randint(1, sides)

print(roll_dice())
print(roll_dice(20))
```

```python
import random

def toss_coin():
    return random.choice(
        ["Heads", "Tails"]
    )

print(toss_coin())
```

```python
import random

def generate_name():

    first = [
        "Shadow",
        "Iron",
        "Storm",
        "Fire"
    ]

    second = [
        "Wolf",
        "Blade",
        "Knight",
        "Dragon"
    ]

    return (
        random.choice(first)
        + " "
        + random.choice(second)
    )

print(generate_name())
```

```python
import random

def random_walk(steps):

    position = 0

    for _ in range(steps):

        position += random.choice(
            [-1, 1]
        )

    return position

print(random_walk(100))
```

```python
import random

def generate_story():

    heroes = [
        "wizard",
        "knight",
        "robot"
    ]

    places = [
        "forest",
        "castle",
        "moon"
    ]

    actions = [
        "fought",
        "found",
        "escaped"
    ]

    return (
        f"A {random.choice(heroes)} "
        f"{random.choice(actions)} "
        f"in a {random.choice(places)}."
    )

print(generate_story())
```
