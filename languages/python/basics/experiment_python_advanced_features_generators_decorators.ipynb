```python
def sliding_window(data, size):

    for i in range(len(data) - size + 1):
        yield data[i:i + size]

for window in sliding_window(
    [1,2,3,4,5],
    3
):
    print(window)
```

```python
def batch_generator(data, batch_size):

    for i in range(
        0,
        len(data),
        batch_size
    ):
        yield data[i:i+batch_size]

for batch in batch_generator(
    list(range(20)),
    4
):
    print(batch)
```

```python
registry = {}

def register(name):

    def decorator(func):
        registry[name] = func
        return func

    return decorator

@register("hello")
def greet():
    return "Hello"

print(registry["hello"]())
```

```python
def read_large_file(filename):

    with open(filename, "r") as file:

        for line in file:
            yield line.strip()
```

```python
import time

def retry(func, retries=5):

    delay = 1

    for attempt in range(retries):

        try:
            return func()

        except Exception:

            print(
                f"Retrying in {delay}s..."
            )

            time.sleep(delay)

            delay *= 2
```
