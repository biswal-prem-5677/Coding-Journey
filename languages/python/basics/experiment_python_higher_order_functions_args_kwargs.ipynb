```python
def multiplier(x):

    def multiply(y):
        return x * y

    return multiply

double = multiplier(2)
triple = multiplier(3)

print(double(10))
print(triple(10))
```

```python
def average(*numbers):
    return sum(numbers) / len(numbers)

print(average(10, 20, 30, 40))
```

```python
def student_info(**kwargs):

    for key, value in kwargs.items():
        print(f"{key}: {value}")

student_info(
    name="Priyabrata",
    branch="CSE",
    year=2
)
```

```python
def add_five(x):
    return x + 5

def square(x):
    return x * x

def compose(f, g):
    return lambda x: f(g(x))

result = compose(square, add_five)

print(result(5))
```
