1. WAP to print sum of digits of a number.


```python
n=int(input("Enter a 3 digit number"))
sum=0
while(n > 0):
        remainder=n%10
        sum+=remainder
        n=n//10
print(sum)
```

2. WAP to check whether a number is palindrome or not.

```python
n=int(input("Enter a number"))
r=0
temp=n
while(temp>0):
  r=r*10+(temp%10)
  temp=temp//10
if(r==n):
  print("Palindrome")
else:
  print("Not Pallindrome")
```

3. WAP to find fibonacci series.

```python
terms=int(input("Enter no of terms"))
a,b=0,1
count=0
if(terms<=0):
  print("Enter a positive number")
elif(terms==1):
  print("fibonacci series : ",b)
else:
  print("fibonacci series")
  while(count<terms):
    print(a)
    c=a+b
    a=b
    b=c
    count+=1
```

5. WAP to determine whether a number is Amstrong number or not.

```python
n=int(input("Enter a number"))
sum=0
temp=n
while(temp>0):
  sum=sum+((temp%10)**3)
  temp=temp//10
if(sum==n):
  print("Amstrong Number")
else:
  print("Not Amstrong Number")
```
