```python
import time

cache = {}

def cached(func):

    def wrapper(x):

        now = time.time()

        if x in cache:

            value, timestamp = cache[x]

            if now - timestamp < 60:
                return value

        result = func(x)

        cache[x] = (
            result,
            now
        )

        return result

    return wrapper
```

```python
events = {}

def subscribe(event, callback):

    if event not in events:
        events[event] = []

    events[event].append(callback)

def emit(event):

    for callback in events.get(event, []):
        callback()

subscribe(
    "login",
    lambda: print("User logged in")
)

emit("login")
```

```python
class Database:

    def save(self):
        print("Saved")

def create_user(db):

    db.save()

database = Database()

create_user(database)
```

```python
from concurrent.futures import ThreadPoolExecutor

def square(n):
    return n * n

with ThreadPoolExecutor() as executor:

    results = list(
        executor.map(
            square,
            range(10)
        )
    )

print(results)
```

```python
from functools import cache

@cache
def expensive_operation(n):

    print("Calculating...")

    return n * n

print(expensive_operation(10))
print(expensive_operation(10))
```
