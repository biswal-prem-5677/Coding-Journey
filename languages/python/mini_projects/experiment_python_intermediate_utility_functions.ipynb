```python
def create_id_generator():

    current = 0

    def next_id():

        nonlocal current

        current += 1

        return current

    return next_id


generate = create_id_generator()

print(generate())
print(generate())
print(generate())
```

```python
def group_by(items, key):

    result = {}

    for item in items:

        group = item[key]

        result.setdefault(
            group,
            []
        ).append(item)

    return result


students = [
    {"name": "A", "branch": "CSE"},
    {"name": "B", "branch": "ECE"},
    {"name": "C", "branch": "CSE"}
]

print(
    group_by(
        students,
        "branch"
    )
)
```

```python
def flatten_dict(
    data,
    parent=""
):

    result = {}

    for key, value in data.items():

        new_key = (
            f"{parent}.{key}"
            if parent
            else key
        )

        if isinstance(value, dict):

            result.update(
                flatten_dict(
                    value,
                    new_key
                )
            )

        else:
            result[new_key] = value

    return result
```

```python
def moving_average(
    numbers,
    window
):

    result = []

    for i in range(
        len(numbers) - window + 1
    ):

        avg = sum(
            numbers[i:i+window]
        ) / window

        result.append(avg)

    return result


print(
    moving_average(
        [10,20,30,40,50],
        3
    )
)
```

```python
def mask_password(password):

    if len(password) <= 4:
        return "*" * len(password)

    return (
        "*" *
        (len(password)-4)
        +
        password[-4:]
    )


print(
    mask_password(
        "supersecret123"
    )
)
```
