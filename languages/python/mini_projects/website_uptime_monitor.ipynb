```python
import requests

def check_url(url):
    try:
        response = requests.get(url)
        print(url, "Status:", response.status_code)
    except:
        print(url, "is unreachable")


urls = [
    "https://google.com",
    "https://github.com",
    "https://openai.com"
]

for url in urls:
    check_url(url)
```

```python
import requests
import time
import logging

logging.basicConfig(
    filename="uptime_log.txt",
    level=logging.INFO
)

urls = [
    "https://google.com",
    "https://github.com",
    "https://openai.com"
]

# Change the 'while True' to a loop with a bounding condition, e.g., 'for i in range(5)'
# This will run the check 5 times.
for i in range(2):

    for url in urls:

        try:

            response = requests.get(url, timeout=5)

            status = response.status_code

            print(url, status)

            logging.info(f"{url} {status}")

        except Exception as e:

            print(url, "DOWN")

            logging.error(f"{url} DOWN")

    time.sleep(60)
print("Monitoring complete.")
```
