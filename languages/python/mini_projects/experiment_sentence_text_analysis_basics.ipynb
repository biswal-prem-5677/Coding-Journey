```python
import string

def analyze_sentence(sentence):
    # Remove punctuation for accurate letter counting
    translator = str.maketrans('', '', string.punctuation)
    clean_sentence = sentence.translate(translator)

    # Split the clean sentence into words
    words = clean_sentence.split()

    # 1. Count total number of words
    num_words = len(words)

    # 2. Count total number of letters
    # Join words back without spaces to count only letters
    total_letters = len(''.join(words))

    # 3. Count letters in each word
    letters_per_word = {word: len(word) for word in words}

    return num_words, total_letters, letters_per_word
```

### How to Use the `analyze_sentence` Function

This function `analyze_sentence(sentence)` takes a string (your sentence) as input and returns three values:

1.  **`num_words`**: The total count of words in the sentence.
2.  **`total_letters`**: The total count of letters in the sentence (punctuation is ignored).
3.  **`letters_per_word`**: A dictionary where keys are the words and values are the number of letters in each respective word.

```python
# Example usage:
sentence = "Hello, world! This is a test sentence."
num_words, total_letters, letters_per_word = analyze_sentence(sentence)

print(f"Original Sentence: '{sentence}'")
print(f"Number of words: {num_words}")
print(f"Total number of letters: {total_letters}")
print(f"Letters per word: {letters_per_word}")

print("\n--- Another Example ---")
sentence_2 = "Python is fun and powerful!"
num_words_2, total_letters_2, letters_per_word_2 = analyze_sentence(sentence_2)

print(f"Original Sentence: '{sentence_2}'")
print(f"Number of words: {num_words_2}")
print(f"Total number of letters: {total_letters_2}")
print(f"Letters per word: {letters_per_word_2}")
```
