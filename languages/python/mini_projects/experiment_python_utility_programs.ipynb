```python
def pyramid(height):

    for i in range(1, height + 1):

        print(
            " " * (height - i)
            + "*" * (2*i - 1)
        )

pyramid(5)
```

```python
import time

def type_writer(text):

    for char in text:

        print(
            char,
            end="",
            flush=True
        )

        time.sleep(0.05)

type_writer("Hello World")
```

```python
import random
import string

def generate_password(length=12):

    chars = (
        string.ascii_letters
        + string.digits
        + "!@#$%"
    )

    return "".join(
        random.choice(chars)
        for _ in range(length)
    )

print(generate_password())
```

```python
import time
from IPython.display import clear_output

def clock():
    try:
        while True:
            # Clear the previous printed output
            clear_output(wait=True)

            # Print the current time
            print(time.strftime("%H:%M:%S"))

            # Wait for 1 second
            time.sleep(1)

    except KeyboardInterrupt:
        print("\nClock stopped.")

clock()
```

```python
import time
import os
from IPython.display import clear_output

# Force Google Colab to use India Standard Time (IST)
os.environ['TZ'] = 'Asia/Kolkata'
time.tzset()

def clock():
    try:
        while True:
            clear_output(wait=True)
            # This will now display your local time
            print("Current Indian Local Time:", time.strftime("%H:%M:%S"))
            time.sleep(1)

    except KeyboardInterrupt:
        print("\nClock stopped by user.")

clock()
```

```python
import random

def terrain(width):

    return [
        random.randint(0, 9)
        for _ in range(width)
    ]

print(terrain(20))
```
