# 🌍 Country Guessing Game 🗺️

Guess the country based on its flag!

````python
import random
from IPython.display import display, Markdown, HTML

# Base URL for flag images from a public CDN
FLAG_CDN_BASE_URL = "https://flagcdn.com/w160/"

# --- Game Resources (You can expand this to 50 countries!) ---
# Each country now includes 'iso_code' for generating flag image URLs.
# You can find ISO 3166-1 alpha-2 codes on Wikipedia or other online resources.
countries = [
    {"name": "United States", "flag_emoji": "🇺🇸", "iso_code": "us"},
    {"name": "Canada", "flag_emoji": "🇨🇦", "iso_code": "ca"},
    {"name": "Mexico", "flag_emoji": "🇲🇽", "iso_code": "mx"},
    {"name": "Brazil", "flag_emoji": "🇧🇷", "iso_code": "br"},
    {"name": "Argentina", "flag_emoji": "🇦🇷", "iso_code": "ar"},
    {"name": "United Kingdom", "flag_emoji": "🇬🇧", "iso_code": "gb"},
    {"name": "France", "flag_emoji": "🇫🇷", "iso_code": "fr"},
    {"name": "Germany", "flag_emoji": "🇩🇪", "iso_code": "de"},
    {"name": "Italy", "flag_emoji": "🇮🇹", "iso_code": "it"},
    {"name": "Spain", "flag_emoji": "🇪🇸", "iso_code": "es"},
    {"name": "Japan", "flag_emoji": "🇯🇵", "iso_code": "jp"},
    {"name": "China", "flag_emoji": "🇨🇳", "iso_code": "cn"},
    {"name": "India", "flag_emoji": "🇮🇳", "iso_code": "in"},
    {"name": "Australia", "flag_emoji": "🇦🇺", "iso_code": "au"},
    {"name": "South Africa", "flag_emoji": "🇿🇦", "iso_code": "za"},
    {"name": "Egypt", "flag_emoji": "🇪🇬", "iso_code": "eg"},
    {"name": "Nigeria", "flag_emoji": "🇳🇬", "iso_code": "ng"},
    {"name": "Russia", "flag_emoji": "🇷🇺", "iso_code": "ru"},
    {"name": "South Korea", "flag_emoji": "🇰🇷", "iso_code": "kr"},
    {"name": "Sweden", "flag_emoji": "🇸🇪", "iso_code": "se"},
    {"name": "Norway", "flag_emoji": "🇳🇴", "iso_code": "no"},
    {"name": "Finland", "flag_emoji": "🇫🇮", "iso_code": "fi"},
    {"name": "Netherlands", "flag_emoji": "🇳🇱", "iso_code": "nl"},
    {"name": "Belgium", "flag_emoji": "🇧🇪", "iso_code": "be"},
    {"name": "Switzerland", "flag_emoji": "🇨🇭", "iso_code": "ch"}
    # Add more countries here to reach 50!
    # Example: {"name": "Ghana", "flag_emoji": "🇬🇭", "iso_code": "gh"}
]

# Dynamically add flag_url to each country based on its iso_code
for country in countries:
    country['flag_url'] = f"{FLAG_CDN_BASE_URL}{country['iso_code']}.png"

if len(countries) < 50:
    display(Markdown(f"**Note:** You currently have {len(countries)} countries. Consider adding more to reach 50 for a fuller game experience! Remember to add the `iso_code` for each new country to enable flag images."))
````

````python
def play_round(countries_list, score):
    if not countries_list:
        display(Markdown("No more countries left to guess! Well played!"))
        return score, False

    selected_country = random.choice(countries_list)
    correct_name = selected_country['name']
    flag_url = selected_country['flag_url'] # Use the image URL

    display(Markdown(f"## What country does this flag belong to?"))
    display(HTML(f"<img src='{flag_url}' alt='{correct_name} flag' width='160px'>")) # Display the image

    user_guess = input("Your guess: ").strip()

    if user_guess.lower() == correct_name.lower():
        score += 1
        display(Markdown(f"**Correct!** Your score: {score}"))
        return score, True
    else:
        display(Markdown(f"**Incorrect!** The correct answer was **{correct_name}**. Your score: {score}"))
        return score, True

def start_game():
    game_score = 0
    remaining_countries = list(countries) # Make a copy to modify

    display(Markdown("# Let's begin the game!\nType 'quit' at any time to end the game."))

    while True:
        if not remaining_countries:
            display(Markdown(f"**You've guessed all available countries!**\nYour final score: {game_score}"))
            break

        current_country = random.choice(remaining_countries)
        correct_name = current_country['name']
        flag_url = current_country['flag_url'] # Use the image URL

        display(Markdown(f"## Guess the country:"))
        display(HTML(f"<img src='{flag_url}' alt='{correct_name} flag' width='160px'>")) # Display the image
        user_guess = input("Your guess: ").strip()

        if user_guess.lower() == 'quit':
            display(Markdown(f"Game ended. Your final score: {game_score}"))
            break

        if user_guess.lower() == correct_name.lower():
            game_score += 1
            display(Markdown(f"**Correct!** Your score: {game_score}"))
            remaining_countries.remove(current_country) # Remove guessed country
        else:
            display(Markdown(f"**Incorrect!** The correct answer was **{correct_name}**. Your score: {game_score}"))

    display(Markdown("Thanks for playing!"))

# Start the game
start_game()
````
