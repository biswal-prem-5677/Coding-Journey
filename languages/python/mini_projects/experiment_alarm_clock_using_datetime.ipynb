## Alarm Function

This function will set an alarm that triggers at a specified hour and minute. When the time is reached, it will print an 'Alarm!' message.

```python
import datetime
import time

def set_alarm(alarm_hour, alarm_minute):
    """
    Sets an alarm to go off at a specified hour and minute.

    Args:
        alarm_hour (int): The hour for the alarm (24-hour format, 0-23).
        alarm_minute (int): The minute for the alarm (0-59).
    """
    print(f"Alarm set for {alarm_hour:02d}:{alarm_minute:02d}. Waiting...")

    while True:
        current_time = datetime.datetime.now()
        if current_time.hour == alarm_hour and current_time.minute == alarm_minute:
            print("Alarm! Time to wake up!")
            # You could add sound playing functionality here, e.g., using playsound library
            break
        time.sleep(1) # Check every minute to reduce CPU usage
```

### How to use the alarm function

To use the alarm, simply call the `set_alarm` function with the desired hour and minute. For example, to set an alarm for 10:30 AM, you would use `set_alarm(10, 30)`.

**Note:** For the alarm to trigger, you need to run the cell containing the `set_alarm` call and keep the notebook running until the specified time.

```python
# Example usage:
# Set an alarm for the next minute from now (approximately)
# You might need to adjust the minute to be a few minutes in the future
# to give you time to run the cell.

now = datetime.datetime.now()
alarm_h = now.hour
alarm_m = (now.minute + 1) % 60 # Set for the next minute

if alarm_m == 0: # If next minute wraps around to a new hour
    alarm_h = (alarm_h + 1) % 24

set_alarm(alarm_h, alarm_m)
```
