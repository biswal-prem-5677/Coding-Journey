```python
import os
import shutil

source = "/content/sample_data"


if not os.path.exists(source):
    os.makedirs(source)
    print(f"Created directory: {source}. Please upload your files here or change the 'source' variable to an existing directory.")

folders = {
    "Images": [".png", ".jpg", ".jpeg"],
    "Documents": [".pdf", ".docx", ".txt"],
    "Videos": [".mp4", ".mkv"],
    "Code": [".py", ".java", ".cpp"]
}

# Only proceed if the source directory is not empty
if os.listdir(source):
    for file in os.listdir(source):
        name, ext = os.path.splitext(file)

        for folder, extensions in folders.items():
            if ext in extensions:
                folder_path = os.path.join(source, folder)

                if not os.path.exists(folder_path):
                    os.makedirs(folder_path)

                shutil.move(
                    os.path.join(source, file),
                    os.path.join(folder_path, file)
                )
    print("Files organized successfully!")
else:
    print(f"The directory '{source}' is empty. Please add files to it before running the script.")
```

```python
import os
import shutil

source_folder = "downloads"

# Create the source folder if it doesn't exist
os.makedirs(source_folder, exist_ok=True)

file_types = {
    "Images": [".jpg", ".png", ".jpeg", ".gif"],
    "Documents": [".pdf", ".docx", ".txt"],
    "Videos": [".mp4", ".mkv"],
    "Code": [".py", ".java", ".cpp"],
    "Archives": [".zip", ".rar"]
}

for file in os.listdir(source_folder):

    file_path = os.path.join(source_folder, file)

    if os.path.isfile(file_path):

        name, ext = os.path.splitext(file)

        for folder, extensions in file_types.items():

            if ext.lower() in extensions:

                folder_path = os.path.join(source_folder, folder)

                os.makedirs(folder_path, exist_ok=True)

                shutil.move(file_path, os.path.join(folder_path, file))

                print(f"Moved {file} -> {folder}")
```
