## Letter Identifier Game

This game will ask you to enter a character, and then it will identify whether it's a small letter, a capital letter, or a special symbol/sign.

```python
import sys

def identify_character(char):
    if 'a' <= char <= 'z':
        return "a small letter"
    elif 'A' <= char <= 'Z':
        return "a capital letter"
    elif char.isnumeric():
        return "a digit"
    elif char in '!@#$%^&*()_+-=[]{}|;:\'",.<>/?`~':
        return f"a special symbol/sign: {char}"
    else:
        return "an unknown character or symbol"

print("Welcome to the Letter Identifier Game!")
print("Enter a character to identify its type. Type 'exit' to quit.")

while True:
    user_input = input("\nEnter a character: ")

    if user_input.lower() == 'exit':
        print("Thanks for playing! Goodbye!")
        break
    elif len(user_input) != 1:
        print("Please enter exactly one character.")
        continue

    result = identify_character(user_input)
    print(f"The character '{user_input}' is {result}.")
```
