```python
import random

def play_multiplication_game():
    score = 0
    num_questions = 5

    print("Welcome to the Multiplication Game!")
    print(f"Let's practice your multiplication. You'll get {num_questions} questions.")
    print("Type 'exit' or 'q' at any time to quit.")

    for i in range(num_questions):
        factor1 = random.randint(1, 25)  # Random factor from 1 to 25
        factor2 = random.randint(1, 10)  # Random factor from 1 to 10
        correct_answer = factor1 * factor2

        print(f"\nQuestion {i + 1}: What is {factor1} x {factor2}?")
        user_input = input("Your answer: ")

        if user_input.lower() in ['exit', 'q']:
            print("Thanks for playing! Goodbye!")
            break

        try:
            user_answer = int(user_input)
            if user_answer == correct_answer:
                print("Correct!")
                score += 1
            else:
                print(f"Wrong! The correct answer was {correct_answer}.")
        except ValueError:
            print("That's not a valid number! Skipping this question.")

    else: # This block executes if the loop completes normally (not broken by 'exit'/'q')
        print(f"\nGame Over! You got {score} out of {num_questions} questions correct.")
        if score == num_questions:
            print("Amazing! You're a multiplication master!")
        elif score >= num_questions / 2:
            print("Good job! Keep practicing!")
        else:
            print("Keep trying! You'll get better with practice!")

# Run the game
play_multiplication_game()
```
