# 🗺️ The Colab Dungeon Crawler! 🚀

Welcome, young adventurer! In this game, you'll explore a mysterious dungeon filled with exciting challenges.

Your goal is to find the **treasure chest 💰** to win! Beware of scary **monsters 👾** and tricky **traps 🪤**. Both monsters and traps will reduce your **health ❤️** by 1. If your health drops to 0, it's game over!

Finding treasure will also award you **100 points ⭐**.

Let's get ready to play! Here's how you'll move:
- Type `w` to move **up**
- Type `s` to move **down**
- Type `a` to move **left**
- Type `d` to move **right**

**Important:** After typing your move (`w`, `a`, `s`, or `d`), **press `Enter`**. The game will then update and show the new dungeon state. The dungeon elements (treasure, monsters, traps) do not move once the game starts, so plan your path carefully! There are no limits to the number of moves you can make, only to your health.

Good luck!

```python
import random
import time
from IPython.display import clear_output

# Game board settings
MAP_WIDTH = 10
MAP_HEIGHT = 5
NUM_TREASURES = 1 # New: Number of treasures to place

# Player symbol
PLAYER = '🤠'

# Dungeon elements
TREASURE = '💰'
MONSTER = '👾'
TRAP = '🪤'
EMPTY = '⬜'

# Game state variables
player_x, player_y = 0, 0
player_health = 3 # New: Player health
score = 0 # New: Player score

# Function to generate the dungeon map
def generate_dungeon():
    random.seed() # Ensures a new seed is used for truly unique layouts each time
    dungeon = [[EMPTY for _ in range(MAP_WIDTH)] for _ in range(MAP_HEIGHT)]

    # Place treasures
    for _ in range(NUM_TREASURES):
        while True:
            treasure_x, treasure_y = random.randint(0, MAP_WIDTH - 1), random.randint(0, MAP_HEIGHT - 1)
            if dungeon[treasure_y][treasure_x] == EMPTY:
                dungeon[treasure_y][treasure_x] = TREASURE
                break

    # Place monsters (up to 3)
    for _ in range(random.randint(1, 3)):
        while True:
            m_x, m_y = random.randint(0, MAP_WIDTH - 1), random.randint(0, MAP_HEIGHT - 1)
            if dungeon[m_y][m_x] == EMPTY:
                dungeon[m_y][m_x] = MONSTER
                break

    # Place traps (up to 2)
    for _ in range(random.randint(0, 2)):
        while True:
            t_x, t_y = random.randint(0, MAP_WIDTH - 1), random.randint(0, MAP_HEIGHT - 1)
            if dungeon[t_y][t_x] == EMPTY:
                dungeon[t_y][t_x] = TRAP
                break

    return dungeon

# Function to draw the dungeon on the screen
def draw_dungeon():
    clear_output(wait=True) # Clear previous output for a smooth animation
    print("\n--- COLAB DUNGEON CRAWLER ---")
    print(f"Health: {player_health} ❤️ | Score: {score} ⭐\n") # Display health and score
    for y in range(MAP_HEIGHT):
        row_display = []
        for x in range(MAP_WIDTH):
            if x == player_x and y == player_y:
                row_display.append(PLAYER) # Draw player at current position
            else:
                row_display.append(dungeon_map[y][x]) # Draw dungeon element
        print("".join(row_display))
    print("\n---------------------------\n")

def handle_move(move):
    global player_x, player_y, game_over, dungeon_map, player_health, score

    new_x, new_y = player_x, player_y

    if move == 'w':
        new_y -= 1
    elif move == 's':
        new_y += 1
    elif move == 'a':
        new_x -= 1
    elif move == 'd':
        new_x += 1
    else:
        print("Invalid move! Use 'w', 's', 'a', or 'd'.")
        return

    # Check for boundaries
    if 0 <= new_x < MAP_WIDTH and 0 <= new_y < MAP_HEIGHT:
        player_x, player_y = new_x, new_y

        # Check what's at the new position
        current_tile = dungeon_map[player_y][player_x]

        if current_tile == TREASURE:
            print("🎉 You found the TREASURE! You win! 🎉")
            score += 100 # Gain score for treasure
            game_over = True
        elif current_tile == MONSTER:
            print("👻 Oh no! A MONSTER attacked you! You lost 1 health! 👻")
            player_health -= 1
            if player_health <= 0:
                print("Your health is 0! Game Over!")
                game_over = True
        elif current_tile == TRAP:
            print("💥 Ouch! You stepped on a TRAP! You lost 1 health! 💥")
            player_health -= 1
            if player_health <= 0:
                print("Your health is 0! Game Over!")
                game_over = True

        # Clear the current tile to show player moved (unless it's a special item)
        if not game_over: # Only replace with EMPTY if game is not over
             dungeon_map[player_y][player_x] = EMPTY

    else:
        print("Ouch! You hit the dungeon wall!")

# Main game loop
def game_loop():
    global game_over, player_x, player_y, dungeon_map, player_health, score

    # Display instructions at the start of the game
    print("# 🗺️ The Colab Dungeon Crawler! 🚀\n")
    print("Welcome, young adventurer! In this game, you'll explore a mysterious dungeon filled with exciting challenges.\n")
    print("Your goal is to find the **treasure chest 💰** to win! Beware of scary **monsters 👾** and tricky **traps 🪤**. Both monsters and traps will reduce your **health ❤️** by 1. If your health drops to 0, it's game over!\n")
    print("Finding treasure will also award you **100 points ⭐**.\n")
    print("Let's get ready to play! Here's how you'll move:\n")
    print("- Type `w` to move **up**")
    print("- Type `s` to move **down**")
    print("- Type `a` to move **left**")
    print("- Type `d` to move **right**\n")
    print("**Important:** After typing your move (`w`, `a`, `s`, or `d`), **press `Enter`**. The game will then update and show the new dungeon state. The dungeon elements (treasure, monsters, traps) do not move once the game starts, so plan your path carefully! There are no limits to the number of moves you can make, only to your health.\n")
    print("Good luck!\n")

    while True:
        game_over = False
        player_x, player_y = 0, 0
        player_health = 3 # Reset health for new game
        score = 0 # Reset score for new game
        dungeon_map = generate_dungeon() # Generate a new dungeon for each round

        while not game_over:
            draw_dungeon()
            move = input("Enter your move (w/s/a/d): ").lower()
            handle_move(move)
            # time.sleep(0.5) # Removed small delay for smoother feel

        print("\n*** GAME ENDED ***")
        print(f"You finished with a score of: {score} ⭐ and {player_health} ❤️ health!") # Display final score
        play_again = input("Do you want to play again? (yes/no): ").lower()
        if play_again != 'yes':
            break

# Start the game
game_loop()
```
