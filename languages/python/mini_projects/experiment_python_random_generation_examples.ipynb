```python
import random

def generate_planet():

    size = random.randint(
        1000,
        10000
    )

    moons = random.randint(
        0,
        10
    )

    return {
        "size": size,
        "moons": moons
    }

print(generate_planet())
```

```python
import random

def random_color():

    return (
        "#{:06x}"
        .format(
            random.randint(
                0,
                0xFFFFFF
            )
        )
    )

print(random_color())
```

```python
import random

def weather():

    return random.choice([
        "Sunny",
        "Rainy",
        "Cloudy",
        "Stormy"
    ])

print(weather())
```

```python
import random

def create_character():

    return {
        "strength":
            random.randint(1,10),

        "agility":
            random.randint(1,10),

        "intelligence":
            random.randint(1,10)
    }

print(create_character())
```

```python
import random

def ecosystem():

    rabbits = 100

    rabbits += random.randint(
        -20,
        30
    )

    return rabbits

print(ecosystem())
```
