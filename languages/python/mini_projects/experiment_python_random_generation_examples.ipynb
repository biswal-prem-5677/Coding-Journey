```python
import random

def lottery():

    return random.sample(
        range(1, 50),
        6
    )

print(lottery())
```

```python
import random

def npc_talk():

    phrases = [
        "Beware the forest.",
        "I lost my sword.",
        "The king needs help."
    ]

    return random.choice(
        phrases
    )

print(npc_talk())
```

```python
import random

def create_world(size):

    return [
        random.choice([0,1])
        for _ in range(size)
    ]

print(create_world(20))
```

```python
import random

def melody(length):

    notes = [
        "C","D","E",
        "F","G","A","B"
    ]

    return [
        random.choice(notes)
        for _ in range(length)
    ]

print(melody(10))
```

```python
import random

def dna(length):

    bases = [
        "A",
        "T",
        "C",
        "G"
    ]

    return "".join(
        random.choice(bases)
        for _ in range(length)
    )

print(dna(20))
```
